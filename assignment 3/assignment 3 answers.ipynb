# Q2) Conditional Density of the Multivariate Gaussian

Let $\mathbf{Z}$ be the concatenated random matrix
$\mathbf{Z} = \begin{bmatrix} \mathbf{X} \\ \mathbf{Y} \end{bmatrix} \in \mathbb{R}^{n+m}$.


The random variable $\mathbf{Z}$ is said to have a **joint multivariate normal distribution** if every linear combination of its components follows a univariate normal distribution.

For $\mathbf{Z}$ to possess a **joint density function** (meaning its distribution is absolutely continuous with respect to the Lebesgue measure on $\mathbb{R}^{n+m}$), its covariance matrix must be non-singular (positive definite).


The joint density function $f_{\mathbf{X}, \mathbf{Y}}(\mathbf{z})$ for $\mathbf{z} \in \mathbb{R}^{n+m}$ is given by:

$$f_{\mathbf{X}, \mathbf{Y}}(\mathbf{z}) = \frac{1}{\sqrt{(2\pi)^{n+m} |\boldsymbol{\Sigma}|}} \exp\left( -\frac{1}{2} (\mathbf{z} - \boldsymbol{\mu})^T \boldsymbol{\Sigma}^{-1} (\mathbf{z} - \boldsymbol{\mu}) \right).$$

Here:

* **Mean ($\boldsymbol{\mu}$):** $\boldsymbol{\mu} = E[\mathbf{Z}] = \begin{bmatrix} \boldsymbol{\mu}_X \\ \boldsymbol{\mu}_Y \end{bmatrix} \in \mathbb{R}^{n+m}$, where $\boldsymbol{\mu}_X \in \mathbb{R}^n$ and $\boldsymbol{\mu}_Y \in \mathbb{R}^m$.
* **Covariance Matrix ($\boldsymbol{\Sigma}$):** A symmetric, positive definite matrix $\boldsymbol{\Sigma} \in \mathbb{R}^{(n+m) \times (n+m)}$ partitioned as:
$$\boldsymbol{\Sigma} = \begin{bmatrix} \boldsymbol{\Sigma}_{XX} & \boldsymbol{\Sigma}_{XY} \\ \boldsymbol{\Sigma}_{YX} & \boldsymbol{\Sigma}_{YY} \end{bmatrix}$$
    * $\boldsymbol{\Sigma}_{XX} = \text{Var}(\mathbf{X})$ (size $n \times n$)
    * $\boldsymbol{\Sigma}_{YY} = \text{Var}(\mathbf{Y})$ (size $m \times m$)
    * $\boldsymbol{\Sigma}_{XY} = \text{Cov}(\mathbf{X}, \mathbf{Y}) = \boldsymbol{\Sigma}_{YX}^T$ (size $n \times m$)
* **Determinant ($|\boldsymbol{\Sigma}|$):** The normalization factor ensuring the total integral is 1.

1. Show that the conditional distribution is given by $f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y})$

$$f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) = \frac{1}{\sqrt{(2\pi)^n |\mathbf{S}_{XX}|}} \exp\left( -\frac{1}{2} (\mathbf{x} - \boldsymbol{\mu}^*)^T \boldsymbol{\Sigma}_{X\mid Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*) \right),$$
where
* $\boldsymbol{\Sigma}_{X\mid Y} = \boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}$
* $\boldsymbol{\mu}^* = \boldsymbol{\mu}_x + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_y)$

# Conditional Distribution of a Joint Multivariate Normal Random Variable

Let

$$
\mathbf Z =
\begin{bmatrix}
\mathbf X \\
\mathbf Y
\end{bmatrix}
\sim N(\boldsymbol\mu,\boldsymbol\Sigma)
$$

with

$$
\boldsymbol\mu =
\begin{bmatrix}
\boldsymbol\mu_X \\
\boldsymbol\mu_Y
\end{bmatrix}
$$

and covariance matrix

$$
\boldsymbol\Sigma =
\begin{bmatrix}
\boldsymbol\Sigma_{XX} & \boldsymbol\Sigma_{XY} \\
\boldsymbol\Sigma_{YX} & \boldsymbol\Sigma_{YY}
\end{bmatrix}.
$$

The joint density is

$$
f_{X,Y}(x,y)
=
\frac{
1
}{
\sqrt{(2\pi)^{n+m}|\Sigma|}
}
\exp\left(
-\frac12
\begin{bmatrix}
x-\mu_X \\
y-\mu_Y
\end{bmatrix}^T
\Sigma^{-1}
\begin{bmatrix}
x-\mu_X \\
y-\mu_Y
\end{bmatrix}
\right).
$$

By definition of conditional density,

$$
f_{X|Y}(x|y)
=
\frac{
f_{X,Y}(x,y)
}{
f_Y(y)
}.
$$

Since

$$
Y \sim N(\mu_Y,\Sigma_{YY}),
$$

its marginal density is

$$
f_Y(y)
=
\frac{
1
}{
\sqrt{(2\pi)^m |\Sigma_{YY}|}
}
\exp\left(
-\frac12
(y-\mu_Y)^T
\Sigma_{YY}^{-1}
(y-\mu_Y)
\right).
$$

Using the block matrix inverse identity, the quadratic form in the exponent can be rewritten as

$$
Q
=
(x-\mu^*)^T
\Sigma_{X|Y}^{-1}
(x-\mu^*)
+
(y-\mu_Y)^T
\Sigma_{YY}^{-1}
(y-\mu_Y),
$$

where

$$
\mu^*
=
\mu_X
+
\Sigma_{XY}\Sigma_{YY}^{-1}(y-\mu_Y)
$$

and

$$
\Sigma_{X|Y}
=
\Sigma_{XX}
-
\Sigma_{XY}\Sigma_{YY}^{-1}\Sigma_{YX}.
$$

Substituting into the joint density gives

$$
f_{X,Y}(x,y)
=
C
\exp\left(
-\frac12
(x-\mu^*)^T
\Sigma_{X|Y}^{-1}
(x-\mu^*)
\right)
$$

$$
\times
\exp\left(
-\frac12
(y-\mu_Y)^T
\Sigma_{YY}^{-1}
(y-\mu_Y)
\right),
$$

where C is a normalization constant.

Dividing by the marginal density of Y cancels all terms depending only on y, giving

$$
f_{X|Y}(x|y)
=
\frac{
1
}{
\sqrt{(2\pi)^n |\Sigma_{X|Y}|}
}
\exp\left(
-\frac12
(x-\mu^*)^T
\Sigma_{X|Y}^{-1}
(x-\mu^*)
\right).
$$

Therefore,

$$
X|Y=y
\sim
N(\mu^*,\Sigma_{X|Y}),
$$

where

$$
\mu^*
=
\mu_X
+
\Sigma_{XY}\Sigma_{YY}^{-1}(y-\mu_Y)
$$

and

$$
\Sigma_{X|Y}
=
\Sigma_{XX}
-
\Sigma_{XY}\Sigma_{YY}^{-1}\Sigma_{YX}.
$$

Hence the conditional distribution of a jointly Gaussian random vector is also Gaussian.

# Conditional Expectation and Information in a Gaussian Signal-Noise Model

Let

$$
X \sim \mathscr N(0,1)
$$

and

$$
Q \sim \mathscr N(0,4),
$$

with X and Q independent.

The observed signal is

$$
Y=X+Q.
$$

---

## 1. Joint Distribution

Since Y is a linear combination of independent Gaussian random variables,

$$
Y \sim \mathscr N(0,5).
$$

This follows because:

$$
E[Y]=E[X]+E[Q]=0+0=0
$$

and

$$
\mathrm{Var}(Y)
=
\mathrm{Var}(X)+\mathrm{Var}(Q)
=
1+4
=
5.
$$

Thus:

$$
Y \sim \mathscr N(0,5).
$$

---

The covariance is

$$
\mathrm{Cov}(X,Y)
=
\mathrm{Cov}(X,X+Q)
$$

$$
=
\mathrm{Cov}(X,X)+\mathrm{Cov}(X,Q).
$$

Since X and Q are independent:

$$
\mathrm{Cov}(X,Q)=0.
$$

Therefore:

$$
\mathrm{Cov}(X,Y)=\mathrm{Var}(X)=1.
$$

Hence the joint distribution is:

$$
\begin{bmatrix}
X\\
Y
\end{bmatrix}
\sim
\mathscr N
\left(
\begin{bmatrix}
0\\
0
\end{bmatrix},
\begin{bmatrix}
1 & 1\\
1 & 5
\end{bmatrix}
\right).
$$

---

## 2. Information Geometry

X is not measurable with respect to:

$$
\sigma(Y)
$$

because Y does not uniquely determine X.

Since:

$$
Y=X+Q,
$$

many different combinations of X and Q can produce the same observed value of Y.

For example, if:

$$
Y=5,
$$

then possible explanations include:

- large temperature with moderate load
- moderate temperature with large load
- small temperature with extremely large load

Thus the engineer cannot determine X exactly from Y alone.

---

## 3. Conditional Expectation

For jointly Gaussian variables:

$$
E[X|Y=y]
=
\mu_X
+
\frac{\mathrm{Cov}(X,Y)}{\mathrm{Var}(Y)}
(y-\mu_Y).
$$

Substitute values:

$$
\mu_X=0,
\qquad
\mu_Y=0,
$$

$$
\mathrm{Cov}(X,Y)=1,
\qquad
\mathrm{Var}(Y)=5.
$$

Therefore:

$$
E[X|Y=y]
=
\frac15 y.
$$

Hence:

$$
E[X|Y]
=
\frac15 Y.
$$

---

The coefficient

$$
\frac{\mathrm{Cov}(X,Y)}{\mathrm{Var}(Y)}
=
\frac15
$$

is the signal-to-total-variance ratio:

$$
\frac{\text{signal variance}}{\text{signal variance + noise variance}}
=
\frac{1}{1+4}
=
\frac15.
$$

---

## 4. Verification of the Residual

Define:

$$
\widehat X = E[X|Y]
=
\frac15 Y.
$$

Residual:

$$
Z=X-\widehat X.
$$

Thus:

$$
Z
=
X-\frac15Y.
$$

---

### Mean of Z

$$
E[Z]
=
E[X]-\frac15E[Y]
$$

$$
=
0-\frac15(0)
$$

$$
=0.
$$

Hence:

$$
E[Z]=0.
$$

---

### Orthogonality Principle

Compute:

$$
\mathrm{Cov}(Z,Y)
=
\mathrm{Cov}\left(X-\frac15Y,Y\right).
$$

$$
=
\mathrm{Cov}(X,Y)-\frac15\mathrm{Var}(Y).
$$

Substitute:

$$
=1-\frac15(5)
$$

$$
=1-1
$$

$$
=0.
$$

Therefore:

$$
\mathrm{Cov}(Z,Y)=0.
$$

So the estimation error is orthogonal to the observation.

---

## 5. Differential Entropy

For a Gaussian random variable:

$$
h(X)
=
\frac12\ln(2\pi e\sigma^2).
$$

Since:

$$
\sigma_X^2=1,
$$

we get:

$$
h(X)
=
\frac12\ln(2\pi e).
$$

---

Conditional variance:

$$
\sigma_{X|Y}^2
=
\sigma_X^2
-
\frac{\mathrm{Cov}(X,Y)^2}{\mathrm{Var}(Y)}.
$$

Substitute:

$$
=
1-\frac{1^2}{5}
$$

$$
=
\frac45.
$$

Thus:

$$
h(X|Y)
=
\frac12\ln\left(2\pi e\frac45\right).
$$

---

## 6. Mutual Information

Mutual information:

$$
I(X;Y)
=
h(X)-h(X|Y).
$$

Thus:

$$
I(X;Y)
=
\frac12
\ln\left(
\frac{1}{4/5}
\right).
$$

$$
=
\frac12\ln\left(\frac54\right).
$$

$$
\approx0.1116
\text{ nats}.
$$

---

### Percentage of Uncertainty Removed

Reduction factor:

$$
1-\frac{\sigma_{X|Y}^2}{\sigma_X^2}
$$

$$
=
1-\frac45
$$

$$
=\frac15.
$$

Thus approximately:

$$
20\%
$$

of the uncertainty is removed.

---

### If Noise Variance Were 100

Then:

$$
Y=X+Q,
\qquad
Q\sim\mathscr N(0,100).
$$

Now:

$$
\mathrm{Var}(Y)=101.
$$

Conditional variance becomes:

$$
\sigma_{X|Y}^2
=
1-\frac{1}{101}
=
\frac{100}{101}.
$$

Almost all uncertainty remains.

Mutual information becomes extremely small.

Thus very noisy measurements provide little information about temperature.

---

## 7. Diagnostic Interpretation

The manager suggests using:

$$
\widehat X=Y.
$$

But:

$$
Y=X+Q,
$$

so Y contains both signal and noise.

Its variance is much larger than the variance of X.

The optimal estimator is instead:

$$
E[X|Y]
=
\frac15Y.
$$

This shrinks the observation toward zero because much of Y is caused by mechanical noise rather than temperature itself.

The estimator reduces overreaction to noisy measurements and gives the minimum mean-square estimation error.
