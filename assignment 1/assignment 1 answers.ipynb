# Q1) The Probability of Shooting at a Target

A point is chosen at random from a disk of radius $10$. Let $A$ be the event that the point lies within $1$ unit of the boundary.

1. Model the experiment assuming the point is uniformly distributed with respect to area, and compute $P(A)$.
2. Model the experiment assuming the distance to the point from the center is chosen uniformly from $[0,10]$, and the direction is chosen independently and uniformly from $[0,2\pi)$ and compute $P(A)$.
3. Explain why the two answers differ.


# Probability in a Disk

A point is chosen at random from a disk of radius 10.  
Let A be the event that the point lies within 1 unit of the boundary.

---

## 1. Uniform distribution with respect to area

If the point is uniformly distributed over the area, then probability is proportional to area.

Event A corresponds to the outer ring between radius 9 and 10.

P(A) = (Area of outer ring) / (Area of disk)

P(A) = [pi(10^2) - pi(9^2)] / [pi(10^2)]

P(A) = (100pi - 81pi) / 100pi

P(A) = 19 / 100

**Final Answer:** P(A) = 0.19

---

## 2. Uniform radius and uniform angle

Assume:
- r ~ Uniform(0,10)
- theta ~ Uniform(0,2pi)

Event A corresponds to:

9 <= r <= 10

So,

P(A) = (10 - 9) / (10 - 0)

P(A) = 1 / 10

**Final Answer:** P(A) = 0.1

---

## 3. Explanation of the difference

The two answers differ because the models are different.

Area-uniform model:
- Probability depends on area
- Outer regions have more area
- So points are more likely near the boundary

Radius-uniform model:
- All radii are equally likely
- But outer rings have larger area
- So this model underestimates boundary regions

For a truly uniform distribution over the disk, the radius should not be uniform.
Instead, it should follow:

f(r) proportional to r

---

## Final Results

P(A) = 0.19   (area-uniform)

P(A) = 0.1    (radius-uniform)

# Q2) Bertrand's Paradox

A chord of a circle is chosen “at random.” What is the probability that its length exceeds the side length of the inscribed equilateral triangle?

Show that different mathematically natural models of “random chord” lead to different answers.

# Bertrand's Paradox: Random Chord Problem

A chord of a circle is chosen at random. What is the probability that its length exceeds the side length of the inscribed equilateral triangle?

---

## Key Fact

For a circle of radius R:

Side length of the inscribed equilateral triangle:

$$ s = \sqrt{3} R $$

A chord is longer than this if and only if its distance from the center satisfies:

$$ d < \frac{R}{2} $$

---

## Model 1: Random Endpoints on the Circle

Choose two points uniformly on the circumference.

Chord length:

$$ L = 2R \sin(\theta/2) $$

We want:

$$ 2R \sin(\theta/2) > \sqrt{3} R $$

Divide both sides by R:

$$ 2 \sin(\theta/2) > \sqrt{3} $$

$$ \sin(\theta/2) > \frac{\sqrt{3}}{2} $$

This gives:

$$ \theta > \frac{2\pi}{3} $$

Since theta is uniform in [0, pi]:

$$ P = \frac{\pi - \frac{2\pi}{3}}{\pi} = \frac{1}{3} $$

Final answer:

$$ P = \frac{1}{3} $$

---

## Model 2: Random Radius and Random Point

Choose:
- a random radius
- a random point on it (uniform from 0 to R)

Distance d is uniform in [0, R].

We need:

$$ d < \frac{R}{2} $$

So:

$$ P = \frac{R/2}{R} = \frac{1}{2} $$

Final answer:

$$ P = \frac{1}{2} $$

---

## Model 3: Random Midpoint (Uniform in Area)

Choose the midpoint uniformly in the disk.

We need:

$$ d < \frac{R}{2} $$

Probability is area ratio:

$$ P = \frac{\pi (R/2)^2}{\pi R^2} $$

$$ P = \frac{R^2/4}{R^2} = \frac{1}{4} $$

Final answer:

$$ P = \frac{1}{4} $$

---

## Conclusion

Different methods give different answers:

$$ \frac{1}{3}, \quad \frac{1}{2}, \quad \frac{1}{4} $$

This happens because "random chord" is not uniquely defined.

Each model uses a different probability distribution.

This is known as Bertrand's Paradox.

#Q8) Probability of Being a Girl?

A family has two children. Assume each child is equally likely to be a boy or a girl, independently of the other.

What is the probability that both children are girls, given that **at least one** of them is a girl?

# Two-Child Family Probability

A family has two children. Assume each child is equally likely to be a boy or a girl, independently of the other.

---

## Sample Space

List all equally likely outcomes:

$$ S = \{BB, BG, GB, GG\} $$

Each outcome has probability:

$$ P = \frac{1}{4} $$

---

## 1. Probability of two boys

Only outcome: BB

$$ P(\text{two boys}) = \frac{1}{4} $$

---

## 2. Probability of at least one boy

Outcomes with at least one boy:

$$ \{BB, BG, GB\} $$

So:

$$ P(\text{at least one boy}) = \frac{3}{4} $$

---

## 3. Given at least one boy, probability both are boys

We use conditional probability:

$$ P(BB \mid \text{at least one boy}) $$

Restrict sample space to:

$$ \{BB, BG, GB\} $$

Only one favorable outcome (BB), so:

$$ P = \frac{1}{3} $$

---

## Final Answers

- Probability of two boys = 1/4  
- Probability of at least one boy = 3/4  
- Probability both are boys given at least one boy = 1/3

# Q9) Information Available in a Discrete Random Variable

A factory inspects a product and records its quality using two different coding systems.

Each product falls into exactly one of the following four categories:

* $G_1$: good quality, produced on Machine 1
* $G_2$: good quality, produced on Machine 2
* $D_1$: defective, produced on Machine 1
* $D_2$: defective, produced on Machine 2

Assume the sample space is

$$
\Omega=\{G_1,G_2,D_1,D_2\},
$$

with probabilities

$$
P(\{G_1\})=0.50,\qquad P(\{G_2\})=0.20,\qquad P(\{D_1\})=0.10,\qquad P(\{D_2\})=0.20.
$$

Define two discrete random variables on the same probability space:

* $X$ records only whether the item is good or defective:
  $$
  X(G_1)=0,\quad X(G_2)=0,\quad X(D_1)=1,\quad X(D_2)=1.
  $$

* $Y$ records both quality and machine:
  $$
  Y(G_1)=1,\quad Y(G_2)=2,\quad Y(D_1)=3,\quad Y(D_2)=4.
  $$

Answer the following.

1. Write down the probability space $(\Omega,\mathfrak F,P)$, where $\mathfrak F=\mathfrak P(\Omega)$.

2. Compute the $\sigma$-algebra generated by $X$, namely $\sigma(X)$.

3. Compute the $\sigma$-algebra generated by $Y$, namely $\sigma(Y)$.

4. Show that $\sigma(X)\subseteq \sigma(Y)$, and explain why $Y$ contains more information than $X$.

5. Find the marginal distribution of $X$.

6. Find the marginal distribution of $Y$.

7. Compute the joint probabilities
   $$
   P(X=0,Y=1),\quad P(X=0,Y=2),\quad P(X=1,Y=3),\quad P(X=1,Y=4).
   $$

8. Compute the conditional probabilities
   $$
   P(X=1\mid Y=3),\qquad P(X=1\mid Y=4),\qquad P(Y=3\mid X=1),\qquad P(Y=4\mid X=1).
   $$

9. Interpret the difference between conditioning on $X$ and conditioning on $Y$.

10. Explain, in words, what information is available in $\sigma(X)$ but not in $\sigma(Y)$, and what information is available in $\sigma(Y)$ but not in $\sigma(X)$.

11. Give a Python based visualization of the model.

12. Find the marginal, joint, and conditional Shannon Entropy of $X$ and $Y$.

# Discrete Probability Model with Two Random Variables

---

## 1. Probability Space

Sample space:

$$ \Omega = \{G_1, G_2, D_1, D_2\} $$

Sigma-algebra:

$$ \mathfrak{F} = \mathfrak{P}(\Omega) $$

Probabilities:

$$
P(G_1)=0.50,\quad
P(G_2)=0.20,\quad
P(D_1)=0.10,\quad
P(D_2)=0.20
$$

---

## 2. Sigma-algebra generated by X

X groups outcomes into:

- X = 0: {G_1, G_2}
- X = 1: {D_1, D_2}

So:

$$
\sigma(X) = \{\emptyset,\ \Omega,\ \{G_1,G_2\},\ \{D_1,D_2\}\}
$$

---

## 3. Sigma-algebra generated by Y

Y distinguishes every outcome:

$$
\sigma(Y) = \mathfrak{P}(\Omega)
$$

(all subsets of Omega)

---

## 4. Relation between sigma-algebras

$$
\sigma(X) \subseteq \sigma(Y)
$$

Reason:

- X only tells "good or defective"
- Y tells both "quality and machine"

So Y has strictly more information.

---

## 5. Marginal distribution of X

$$
P(X=0) = P(G_1)+P(G_2) = 0.50 + 0.20 = 0.70
$$

$$
P(X=1) = P(D_1)+P(D_2) = 0.10 + 0.20 = 0.30
$$

---

## 6. Marginal distribution of Y

$$
P(Y=1)=0.50,\quad
P(Y=2)=0.20,\quad
P(Y=3)=0.10,\quad
P(Y=4)=0.20
$$

---

## 7. Joint probabilities

Since Y uniquely determines the outcome:

$$
P(X=0,Y=1)=0.50
$$

$$
P(X=0,Y=2)=0.20
$$

$$
P(X=1,Y=3)=0.10
$$

$$
P(X=1,Y=4)=0.20
$$

---

## 8. Conditional probabilities

$$
P(X=1 \mid Y=3)=1
$$

$$
P(X=1 \mid Y=4)=1
$$

$$
P(Y=3 \mid X=1)=\frac{0.10}{0.30}=\frac{1}{3}
$$

$$
P(Y=4 \mid X=1)=\frac{0.20}{0.30}=\frac{2}{3}
$$

---

## 9. Interpretation

- Conditioning on Y gives exact information about the product.
- Conditioning on X only gives partial information (quality only).

---

## 10. Information comparison

- sigma(X): tells only whether product is good or defective
- sigma(Y): tells both machine and quality

So:

- Y contains strictly more information than X
- X is a "coarse" version of Y

---

## 12. Shannon Entropy

Entropy formula:

$$
H(X) = -\sum p(x)\log_2 p(x)
$$

### Entropy of X

$$
H(X) = -(0.7\log_2 0.7 + 0.3\log_2 0.3)
$$

### Entropy of Y

$$
H(Y) = -(0.5\log_2 0.5 + 0.2\log_2 0.2 + 0.1\log_2 0.1 + 0.2\log_2 0.2)
$$

### Conditional entropy

Since Y determines X:

$$
H(X|Y) = 0
$$

$$
H(Y|X) > 0
$$

```python
import numpy as np
import matplotlib.pyplot as plt

# Probabilities
P = {
    "G1": 0.50,
    "G2": 0.20,
    "D1": 0.10,
    "D2": 0.20
}

# X distribution
PX = {
    0: P["G1"] + P["G2"],
    1: P["D1"] + P["D2"]
}

# Y distribution
PY = {
    1: P["G1"],
    2: P["G2"],
    3: P["D1"],
    4: P["D2"]
}

# Entropy function
def entropy(probs):
    return -sum(p * np.log2(p) for p in probs if p > 0)

HX = entropy(PX.values())
HY = entropy(PY.values())

print("H(X) =", HX)
print("H(Y) =", HY)
print("H(X|Y) = 0")

# Plot distributions
plt.figure()
plt.bar(PX.keys(), PX.values())
plt.title("Distribution of X")
plt.xlabel("X")
plt.ylabel("Probability")
plt.show()

plt.figure()
plt.bar(PY.keys(), PY.values())
plt.title("Distribution of Y")
plt.xlabel("Y")
plt.ylabel("Probability")
plt.show()
```

# Q10) Information Available in Continuous Random Varaibles

A chemical plant monitors each production run using two different measurement systems.

For a given run, let

* $T$ be the deviation of the reactor temperature from target, measured in standardized units,
* $P$ be the deviation of the reactor pressure from target, also measured in standardized units.

Assume that $(T,P)$ is jointly continuous, with independent normal components:

$$
T \sim N(0,1), \qquad P \sim N(0,4),

and $T$ and $P$ are independent.

Let the underlying probability space be

$$
\Omega=\mathbb R^2, \qquad \mathfrak F=\mathfrak B(\mathbb R^2),
$$

and let $P$ be the probability measure induced by the joint density

$$
f_{T,P}(t,p)=\frac{1}{4\pi}\exp\left(-\frac{t^2}{2}-\frac{p^2}{8}\right), \qquad (t,p)\in\mathbb R^2.

Define two random variables on the same probability space:

* $X:\Omega\to\mathbb R$ by
  $$
  X(t,p)=t,
  $$
  so that $X$ records only the reactor temperature deviation;

* $Y:\Omega\to\mathbb R^2$ by
  $$
  Y(t,p)=(t,p),
  $$
  so that $Y$ records the full sensor state: both temperature and pressure deviations.

Answer the following.

2. Compute the $\sigma$-algebra generated by $X$, namely $\sigma(X)$.

3. Compute the $\sigma$-algebra generated by $Y$, namely $\sigma(Y)$.

4. Show that
   $$
   \sigma(X)\subseteq \sigma(Y),
   $$
   and explain why $Y$ contains more information than $X$.

5. Find the marginal density of $X$.

6. Find the marginal density of $Y$.

7. Compute the following probabilities:
   $$
   P(X\leq 0), \qquad P(X>1), \qquad P(Y\in (-\infty,0]\times\mathbb R), \qquad P(Y\in [ -1,1]\times[-2,2]).
   $$

8. Compute the conditional distribution of $X$ given $Y=(t,p)$.

9. Compute the conditional distribution of $Y$ given $X=t$.

10. Explain, in words, what is known when one conditions on $X$ and what additional information becomes available when one conditions on $Y$.

11. Show that
    $$
    X=\pi_1\circ Y,
    $$
    where $\pi_1(t,p)=t$, and explain why this implies that $X$ is a measurable function of $Y$.

12. Give a Python-based visualization of the model, for example by plotting:

* the joint density of $Y=(T,P)$ as a contour plot or heat map;

13. Compute the differential entropy of $X$:

$$
h(X)=-\int_{\mathbb R} f_X(x)\ln f_X(x)\,dx.
$$

14. Compute the differential entropy of $Y$:

$$
h(Y)=-\int_{\mathbb R^2} f_Y(t,p)\ln f_Y(t,p)\,dt\,dp.
$$

15. Compute the conditional differential entropy

$$
h(Y|X),
$$

and interpret it physically.

16. Discuss what happens to

in this model, and explain why the continuous case differs from the discrete case when one random variable is completely determined by the other.


# Continuous Probability Model: Temperature and Pressure

---

## 1. Probability Space

$$ \Omega = \mathbb{R}^2 $$

$$ \mathfrak{F} = \mathfrak{B}(\mathbb{R}^2) $$

Joint density:

$$
f_{T,P}(t,p) = \frac{1}{4\pi} \exp\left(-\frac{t^2}{2} - \frac{p^2}{8}\right)
$$

---

## 2. Sigma-algebra generated by X

Since $X(t,p)=t$:

$$
\sigma(X) = \{ A \times \mathbb{R} : A \in \mathfrak{B}(\mathbb{R}) \}
$$

---

## 3. Sigma-algebra generated by Y

Since $Y(t,p)=(t,p)$ (identity mapping):

$$
\sigma(Y) = \mathfrak{B}(\mathbb{R}^2)
$$

---

## 4. Relation

$$
\sigma(X) \subseteq \sigma(Y)
$$

Reason:
- X observes only temperature
- Y observes temperature and pressure

So Y contains more information.

---

## 5. Marginal density of X

Integrate over p:

$$
f_X(t) = \int_{-\infty}^{\infty} f_{T,P}(t,p)\,dp
$$

$$
f_X(t) = \frac{1}{\sqrt{2\pi}} e^{-t^2/2}
$$

So:

$$
X \sim N(0,1)
$$

---

## 6. Marginal density of Y

Same as joint density:

$$
f_Y(t,p) = f_{T,P}(t,p)
$$

---

## 7. Probabilities

### (a) $$P(X \le 0)$$

By symmetry of normal:

$$
P(X \le 0) = 0.5
$$

---

### (b) $$P(X > 1)$$

$$
P(X > 1) = 1 - \Phi(1) \approx 0.1587
$$

---

### (c) $$P(Y \in (-\infty,0] \times \mathbb{R})$$

Same as:

$$
P(T \le 0) = 0.5
$$

---

### (d) $$P(Y \in [-1,1] \times [-2,2])$$

Using independence:

$$
P = P(-1 \le T \le 1)\, P(-2 \le P \le 2)
$$

$$
= (\Phi(1) - \Phi(-1))(\Phi(1) - \Phi(-1))
$$

$$
\approx (0.6826)^2 \approx 0.466
$$

---

## 8. Conditional distribution of X given Y=(t,p)

Since X is determined by Y:

$$
X = t
$$

So:

- Degenerate distribution at t

---

## 9. Conditional distribution of Y given X=t

Since independence:

$$
P \sim N(0,4)
$$

Thus:

$$
Y | X=t = (t, P), \quad P \sim N(0,4)
$$

---

## 10. Interpretation

- Conditioning on X → only temperature known
- Conditioning on Y → both temperature and pressure known

---

## 11. Functional relationship

Define projection:

$$
\pi_1(t,p) = t
$$

Then:

$$
X = \pi_1 \circ Y
$$

So X is a measurable function of Y.

---

## 13. Differential entropy of X

For normal distribution:

$$
h(X) = \frac{1}{2} \ln(2\pi e)
$$

---

## 14. Differential entropy of Y

For independent normals:

$$
h(Y) = h(T) + h(P)
$$

$$
= \frac{1}{2}\ln(2\pi e) + \frac{1}{2}\ln(8\pi e)
$$

$$
= \ln(4\pi e)
$$

---

## 15. Conditional entropy h(Y|X)

$$
h(Y|X) = h(P) = \frac{1}{2}\ln(8\pi e)
$$

Interpretation:
- Remaining uncertainty is due to pressure only

---

## 16. What about h(X|Y)?

$$
h(X|Y) = -\infty
$$

Reason:
- X is completely determined by Y
- In continuous case, this leads to negative infinite entropy

This differs from discrete case where:

$$
H(X|Y) = 0
$$

---

## Key Insight

- Y fully determines X
- But continuous entropy behaves differently from discrete entropy

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Create grid
t = np.linspace(-4, 4, 100)
p = np.linspace(-8, 8, 100)
T, P = np.meshgrid(t, p)

# Joint density
Z = (1/(4*np.pi)) * np.exp(-T**2/2 - P**2/8)

# Contour plot
plt.figure()
plt.contourf(T, P, Z)
plt.title("Joint Density of (T, P)")
plt.xlabel("T (Temperature)")
plt.ylabel("P (Pressure)")
plt.colorbar()
plt.show()

# Marginal of X
x = np.linspace(-4,4,100)
fx = norm.pdf(x, 0, 1)

plt.figure()
plt.plot(x, fx)
plt.title("Marginal Density of X")
plt.xlabel("X")
plt.ylabel("Density")
plt.show()

# Simple sigma-algebra illustration (text)
print("sigma(X): sets of form A x R (depends only on T)")
print("sigma(Y): all subsets of R^2 (depends on both T and P)")
```
