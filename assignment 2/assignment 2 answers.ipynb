# Emergency Transformer Failure Model

In a certain electricity distribution district, let X denote the number of emergency transformer failures reported in a single day.

The probabilities are:

$$
P(X=0)=0.50
$$

$$
P(X=1)=0.30
$$

$$
P(X=2)=0.15
$$

$$
P(X=3)=0.05
$$

---

## 1. Probability Space

Sample space:

$$
\Omega = \{0,1,2,3\}
$$

Sigma-algebra:

$$
\mathfrak{F} = \mathfrak{P}(\Omega)
$$

Probability measure:

$$
P(\{0\})=0.50
$$

$$
P(\{1\})=0.30
$$

$$
P(\{2\})=0.15
$$

$$
P(\{3\})=0.05
$$

Check:

$$
0.50+0.30+0.15+0.05=1
$$

So this is a valid probability space.

---

## 2. Definition of the Random Variable

Define

$$
X:\Omega \to \mathbb{R}
$$

by

$$
X(\omega)=\omega
$$

Thus X records the number of emergency transformer failures in one day.

---

## 3. Probability Mass Function (PMF)

The probability mass function is

$$
p_X(x)=P(X=x)
$$

given by:

$$
p_X(0)=0.50
$$

$$
p_X(1)=0.30
$$

$$
p_X(2)=0.15
$$

$$
p_X(3)=0.05
$$

---

## 4. Expectation of X

The expectation is

$$
E[X]=\sum xP(X=x)
$$

Therefore,

$$
E[X]
=0(0.50)+1(0.30)+2(0.15)+3(0.05)
$$

$$
=0+0.30+0.30+0.15
$$

$$
=0.75
$$

Final answer:

$$
E[X]=0.75
$$

---

## 5. Variance of X

First compute

$$
E[X^2]
$$

$$
E[X^2]
=0^2(0.50)+1^2(0.30)+2^2(0.15)+3^2(0.05)
$$

$$
=0+0.30+0.60+0.45
$$

$$
=1.35
$$

Variance formula:

$$
\mathrm{Var}(X)=E[X^2]-(E[X])^2
$$

$$
\mathrm{Var}(X)=1.35-(0.75)^2
$$

$$
=1.35-0.5625
$$

$$
=0.7875
$$

Final answer:

$$
\mathrm{Var}(X)=0.7875
$$

---

## 6. Probabilities

### Probability of at least one failure

$$
P(X\ge1)=P(X=1)+P(X=2)+P(X=3)
$$

$$
=0.30+0.15+0.05
$$

$$
=0.50
$$

---

### Probability of at least two failures

$$
P(X\ge2)=P(X=2)+P(X=3)
$$

$$
=0.15+0.05
$$

$$
=0.20
$$

---

## 7. Practical Interpretation

### Expectation

$$
E[X]=0.75
$$

This means that, on average, the district experiences about 0.75 emergency transformer failures per day.

---

### Variance

$$
\mathrm{Var}(X)=0.7875
$$

The variance measures how much the number of failures fluctuates from day to day.

A larger variance would indicate more unpredictable failure behavior.

# Beta Distribution Model for Defective Proportion

Let X denote the proportion of defective items in a large production batch.

Assume:

$$
X \sim \text{Beta}(2,18)
$$

with density

$$
f(x)=
\begin{cases}
\dfrac{1}{B(2,18)}x(1-x)^{17}, & 0<x<1 \\
0, & \text{otherwise}
\end{cases}
$$

---

## 1. Probability Space

Choose

$$
\Omega=[0,1]
$$

Sigma-algebra:

$$
\mathfrak F=\mathfrak B([0,1])
$$

where $\mathfrak B([0,1])$ is the Borel sigma-algebra on $[0,1]$.

Define probability measure:

$$
P(A)=\int_A f(x)\,dx
$$

for all measurable sets $A \subseteq [0,1]$.

---

## 2. Definition of X

Define

$$
X:\Omega \to \mathbb R
$$

by

$$
X(\omega)=\omega
$$

Thus X records the defective proportion.

---

## 3. Probability Density Function

The density function is

$$
f(x)=
\begin{cases}
\dfrac{1}{B(2,18)}x(1-x)^{17}, & 0<x<1 \\
0, & \text{otherwise}
\end{cases}
$$

For Beta distributions:

$$
B(\alpha,\beta)=\frac{\Gamma(\alpha)\Gamma(\beta)}{\Gamma(\alpha+\beta)}
$$

Thus,

$$
B(2,18)=\frac{\Gamma(2)\Gamma(18)}{\Gamma(20)}
$$

Since

$$
\Gamma(2)=1!
$$

and

$$
\Gamma(18)=17!
$$

we get

$$
B(2,18)=\frac{17!}{19!}
=\frac{1}{19 \times 18}
=\frac{1}{342}
$$

Hence:

$$
f(x)=342x(1-x)^{17}
$$

for $0<x<1$.

---

## 4. Verification that f is a Valid Density

A valid density must satisfy:

### Nonnegative

$$
f(x)\ge0
$$

for all x.

This is true because:

- $x\ge0$
- $(1-x)^{17}\ge0$

---

### Total probability equals 1

$$
\int_0^1 f(x)\,dx
=
\frac{1}{B(2,18)}
\int_0^1 x(1-x)^{17}\,dx
$$

By definition of the Beta function:

$$
\int_0^1 x(1-x)^{17}\,dx
=
B(2,18)
$$

Therefore:

$$
\int_0^1 f(x)\,dx=1
$$

So f is a valid probability density function.

---

## 5. Expectation

For a Beta distribution:

$$
E[X]=\frac{\alpha}{\alpha+\beta}
$$

Thus:

$$
E[X]=\frac{2}{20}=0.10
$$

Final answer:

$$
E[X]=0.10
$$

---

## 6. Variance

For a Beta distribution:

$$
\mathrm{Var}(X)
=
\frac{\alpha\beta}
{(\alpha+\beta)^2(\alpha+\beta+1)}
$$

Substitute $\alpha=2$, $\beta=18$:

$$
\mathrm{Var}(X)
=
\frac{2(18)}
{20^2(21)}
$$

$$
=
\frac{36}{8400}
$$

$$
=
0.004286
$$

Final answer:

$$
\mathrm{Var}(X)\approx0.00429
$$

---

## 7. Probability that X ≤ 0.10

$$
P(X\le0.10)
=
\int_0^{0.10} 342x(1-x)^{17}\,dx
$$

Equivalent Beta form:

$$
P(X\le0.10)
=
\frac{1}{B(2,18)}
\int_0^{0.10} x(1-x)^{17}\,dx
$$

---

## 8. Probability that 0.05 ≤ X ≤ 0.15

$$
P(0.05\le X\le0.15)
=
\int_{0.05}^{0.15}342x(1-x)^{17}\,dx
$$

Equivalent Beta form:

$$
P(0.05\le X\le0.15)
=
\frac{1}{B(2,18)}
\int_{0.05}^{0.15}x(1-x)^{17}\,dx
$$

---

## 9. Why Beta Distribution is Appropriate

The Beta distribution is suitable because:

- X represents a proportion
- Proportions must lie between 0 and 1
- Beta distributions are flexible and can model many shapes
- The parameters control the concentration and skewness

Here, most probability mass is near small defect rates, which is realistic in manufacturing quality control.

---

## 10. Interpretation of Expectation and Variance

### Expectation

$$
E[X]=0.10
$$

This means the average defective proportion is about 10%.

---

### Variance

$$
\mathrm{Var}(X)\approx0.00429
$$

This measures how much the defective proportion varies from day to day.

A small variance indicates relatively stable production quality.

# Triangular Distribution Model for Generator Installation Time

Let X denote the number of hours required to install a backup generator.

Assume:

- Minimum possible time: a = 4
- Maximum possible time: b = 10
- Most likely time: c = 6

Thus:

$$
X \sim \text{Triangular}(a=4,b=10,c=6)
$$

---

## 1. Support of X

The support is:

$$
4 \le X \le 10
$$

---

## 2. Probability Density Function

For a triangular distribution:

$$
f_X(x)=
\begin{cases}
\dfrac{2(x-a)}{(b-a)(c-a)}, & a \le x \le c \\[8pt]
\dfrac{2(b-x)}{(b-a)(b-c)}, & c < x \le b \\[8pt]
0, & \text{otherwise}
\end{cases}
$$

Substitute:

$$
a=4,\quad b=10,\quad c=6
$$

For:

$$
4 \le x \le 6
$$

$$
f_X(x)=\frac{2(x-4)}{(10-4)(6-4)}
$$

$$
=\frac{2(x-4)}{12}
$$

$$
=\frac{x-4}{6}
$$

For:

$$
6 < x \le 10
$$

$$
f_X(x)=\frac{2(10-x)}{(10-4)(10-6)}
$$

$$
=\frac{2(10-x)}{24}
$$

$$
=\frac{10-x}{12}
$$

Therefore:

$$
f_X(x)=
\begin{cases}
\dfrac{x-4}{6}, & 4 \le x \le 6 \\[8pt]
\dfrac{10-x}{12}, & 6 < x \le 10 \\[8pt]
0, & \text{otherwise}
\end{cases}
$$

---

We compute:

$$
\int_4^6 \frac{x-4}{6}\,dx
+
\int_6^{10} \frac{10-x}{12}\,dx
$$

First integral:

$$
\int_4^6 \frac{x-4}{6}\,dx
=
\frac{1}{6}\int_4^6 (x-4)\,dx
$$

$$
=
\frac{1}{6}\left[\frac{(x-4)^2}{2}\right]_4^6

$$
=
\frac{1}{6}\times 2
=
\frac{1}{3}
$$

Second integral:

$$
\int_6^{10}\frac{10-x}{12}\,dx
=
\frac{1}{12}\int_6^{10}(10-x)\,dx
$$

$$
=
\frac{1}{12}\times 8
=
\frac{2}{3}
$$

Total:

$$
\frac{1}{3}+\frac{2}{3}=1
$$

Hence f is a valid density.
