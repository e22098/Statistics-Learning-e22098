# Statistical Mechanics of the Quantum Oscillator

## 1. The Partition Function

The canonical partition function $Z(T)$ is given by the sum over all quantum states:

$$Z(T) = \sum_{j=0}^\infty e^{-\beta E_j}$$

where $\beta = \frac{1}{k_B T}$ and $E_j = \hbar \omega \left(j + \frac{1}{2}\right)$. Substituting the energy eigenvalues:

$$Z(T) = \sum_{j=0}^\infty e^{-\beta \hbar \omega \left(j + \frac{1}{2}\right)} = e^{-\frac{\beta \hbar \omega}{2}} \sum_{j=0}^\infty \left(e^{-\beta \hbar \omega}\right)^j$$

This is an infinite geometric series of the form $\sum_{j=0}^\infty x^j$, where the common ratio is $x = e^{-\beta \hbar \omega}$. Since $\beta, \hbar, \omega > 0$, we have $|x| < 1$, meaning the series converges to $\frac{1}{1-x}$.

Applying the geometric series sum formula:

$$Z(T) = e^{-\frac{\beta \hbar \omega}{2}} \cdot \frac{1}{1 - e^{-\beta \hbar \omega}}$$

Substituting $\beta = \frac{1}{k_B T}$ back into the expression yields the desired result:

$$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$$

---

## 2. Thermodynamic Potentials

### Helmholtz Free Energy, $F(T)$
The Helmholtz free energy is defined as $F(T) = -k_B T \ln Z$. Using our simplified expression for $Z$:

$$\ln Z = \ln\left(e^{-\frac{\beta \hbar \omega}{2}}\right) - \ln\left(1 - e^{-\beta \hbar \omega}\right) = -\frac{\beta \hbar \omega}{2} - \ln\left(1 - e^{-\beta \hbar \omega}\right)$$

Multiplying by $-k_B T$ (and noting that $-k_B T \cdot (-\frac{\beta \hbar \omega}{2}) = \frac{\hbar \omega}{2}$):

$$F(T) = \frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_BT}}\right)$$

### Internal Energy, $U(T)$
The internal energy can be computed using the relation $U = -\frac{\partial \ln Z}{\partial \beta}$:

$$U = -\frac{\partial}{\partial \beta} \left[ -\frac{\beta \hbar \omega}{2} - \ln\left(1 - e^{-\beta \hbar \omega}\right) \right]$$

$$U = \frac{\hbar \omega}{2} + \frac{1}{1 - e^{-\beta \hbar \omega}} \cdot \left(-e^{-\beta \hbar \omega}\right) \cdot (-\hbar \omega)$$

$$U = \frac{\hbar \omega}{2} + \frac{\hbar \omega e^{-\beta \hbar \omega}}{1 - e^{-\beta \hbar \omega}}$$

Multiplying the numerator and denominator of the second term by $e^{\beta \hbar \omega}$:

$$U(T) = \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_BT}} - 1}$$

---

## 3. Entropy

The entropy $S(T)$ is obtained via $S = -\left(\frac{\partial F}{\partial T}\right)$. Let us differentiate $F(T)$:

$$S = -\frac{\partial}{\partial T} \left[ \frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_BT}}\right) \right]$$

Using the product rule on the second term:

$$S = -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_BT}}\right) - k_B T \cdot \frac{1}{1 - e^{-\frac{\hbar \omega}{k_BT}}} \cdot \left(-e^{-\frac{\hbar \omega}{k_BT}}\right) \cdot \left(-\frac{\hbar \omega}{k_B T^2}\right)$$

$$S = -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_BT}}\right) + \frac{\hbar \omega}{T} \frac{e^{-\frac{\hbar \omega}{k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$$

$$S(T) = -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_BT}}\right) + \frac{\hbar \omega}{T \left(e^{\frac{\hbar \omega}{k_BT}} - 1\right)}$$

### Verification of $U = F + TS$
Let us check the thermodynamic identity by substituting $F$ and $S$:

$$F + TS = \left[\frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_BT}}\right)\right] + T \left[-k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_BT}}\right) + \frac{\hbar \omega}{T \left(e^{\frac{\hbar \omega}{k_BT}} - 1\right)}\right]$$

$$F + TS = \frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_BT}}\right) - k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_BT}}\right) + \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_BT}} - 1}$$

$$F + TS = \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_BT}} - 1} = U(T)$$

The identity is verified.

---

## 4. Heat Capacity and Hyperbolic Form

The heat capacity at constant volume is defined as $C_V = \frac{\partial U}{\partial T}$. Differentiating our expression for $U(T)$:

$$C_V = \frac{\partial}{\partial T} \left[ \frac{\hbar \omega}{2} + \hbar \omega \left(e^{\frac{\hbar \omega}{k_BT}} - 1\right)^{-1} \right]$$

$$C_V = \hbar \omega \cdot (-1)\left(e^{\frac{\hbar \omega}{k_BT}} - 1\right)^{-2} \cdot e^{\frac{\hbar \omega}{k_BT}} \cdot \left(-\frac{\hbar \omega}{k_B T^2}\right)$$

$$C_V = k_B \left(\frac{\hbar \omega}{k_B T}\right)^2 \frac{e^{\frac{\hbar \omega}{k_BT}}}{\left(e^{\frac{\hbar \omega}{k_BT}} - 1\right)^2}$$

To convert this into hyperbolic functions, let $x = \frac{\hbar \omega}{2k_BT}$. Then $2x = \frac{\hbar \omega}{k_BT}$. Substituting this into the equation:

$$C_V = k_B (2x)^2 \frac{e^{2x}}{\left(e^{2x} - 1\right)^2} = 4k_B x^2 \frac{e^{2x}}{\left[e^x\left(e^x - e^{-x}\right)\right]^2} = 4k_B x^2 \frac{e^{2x}}{e^{2x}\left(e^x - e^{-x}\right)^2}$$

$$C_V = k_B x^2 \frac{4}{\left(e^x - e^{-x}\right)^2} = k_B x^2 \frac{1}{\left(\frac{e^x - e^{-x}}{2}\right)^2}$$

Since $\sinh(x) = \frac{e^x - e^{-x}}{2}$, we arrive at the compact hyperbolic form:

$$C_V = k_B \left(\frac{\hbar \omega}{2k_BT}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)}$$

---

## 5. Asymptotic Behavior (Discussion)

In the high-temperature limit, the thermal energy is much larger than the quantum energy spacing ($k_B T \gg \hbar \omega$). This implies that the dimensionless variable $x = \frac{\hbar \omega}{2k_BT} \ll 1$.

We can apply the Taylor series expansion for the hyperbolic sine function for small arguments, $\sinh(x) \approx x$:

$$C_V \approx k_B x^2 \cdot \frac{1}{x^2} = k_B$$

### Conclusion
Yes, this matches the classical **equipartition theorem** prediction. A classical 1D harmonic oscillator has two quadratic degrees of freedom (kinetic energy $\propto p^2$ and potential energy $\propto x^2$). According to the equipartition theorem, each quadratic degree of freedom contributes $\frac{1}{2}k_B$ to the heat capacity, yielding a total of:

$$C_V^{\text{classical}} = 2 \times \frac{1}{2}k_B = k_B$$

As $T \to \infty$, the quantum mechanical nature of the discrete energy levels becomes negligible, and the system smoothly transitions into classical behavior.

# Q. The Maximum Entropy Distribution under Mean and Variance Constraints

Consider the random variable $X \in (-\infty, \infty)$. Let:

1. **The Mean:** $\mathbb{E}[X] = \mu$
2. **The Variance:** $\mathbb{E}[(X-\mu)^2] = \sigma^2$

Show that the law of $X$ that maximizes entropy is given by:

$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

# The Maximum Entropy Distribution under Mean and Variance Constraints

To find the probability density function $f(x)$ for a continuous random variable $X \in (-\infty, \infty)$ that maximizes the differential entropy under given mean and variance constraints, we use the method of **Lagrange Multipliers** combined with the **Calculus of Variations**.

---

## 1. Objective and Constraints

We aim to maximize the differential entropy functional:

$$h(f) = -\int_{-\infty}^{\infty} f(x) \ln f(x) \, dx$$

Subject to three operational constraints:
1. **Normalization:** $\int_{-\infty}^{\infty} f(x) \, dx = 1$
2. **Mean:** $\int_{-\infty}^{\infty} x f(x) \, dx = \mu$
3. **Variance:** $\int_{-\infty}^{\infty} (x - \mu)^2 f(x) \, dx = \sigma^2$

---

## 2. Setting Up the Lagrangian Functional

We construct the Lagrangian functional $\mathcal{L}(f)$ by introducing the multipliers $\lambda_0$, $\lambda_1$, and $\lambda_2$:

$$\mathcal{L}(f) = -\int_{-\infty}^{\infty} f(x) \ln f(x) \, dx - \lambda_0 \left( \int_{-\infty}^{\infty} f(x) \, dx - 1 \right) - \lambda_1 \left( \int_{-\infty}^{\infty} x f(x) \, dx - \mu \right) - \lambda_2 \left( \int_{-\infty}^{\infty} (x - \mu)^2 f(x) \, dx - \sigma^2 \right)$$

Combining the integrals into a single expression yields:

$$\mathcal{L}(f) = \int_{-\infty}^{\infty} \left[ -f(x) \ln f(x) - \lambda_0 f(x) - \lambda_1 x f(x) - \lambda_2 (x - \mu)^2 f(x) \right] dx + \lambda_0 + \lambda_1 \mu + \lambda_2 \sigma^2$$

---

## 3. Maximizing the Functional

According to the calculus of variations, to maximize $\mathcal{L}(f)$, the functional derivative with respect to $f(x)$ at any point $x$ must equal zero:

$$\frac{\partial}{\partial f} \left( -f(x) \ln f(x) - \lambda_0 f(x) - \lambda_1 x f(x) - \lambda_2 (x - \mu)^2 f(x) \right) = 0$$

Evaluating the derivative yields:

$$- \ln f(x) - 1 - \lambda_0 - \lambda_1 x - \lambda_2 (x - \mu)^2 = 0$$

Isolating $\ln f(x)$:

$$\ln f(x) = -(1 + \lambda_0) - \lambda_1 x - \lambda_2 (x - \mu)^2$$

Taking the exponential of both sides gives the functional form of the optimal probability density:

$$f(x) = \exp\left( -(1 + \lambda_0) - \lambda_1 x - \lambda_2 (x - \mu)^2 \right)$$

---

## 4. Evaluating the Lagrange Multipliers

We rewrite the distribution structure compactly as:

$$f(x) = C \exp\left( -\lambda_1 x - \lambda_2 (x - \mu)^2 \right)$$

where $C = e^{-(1 + \lambda_0)}$ serves as the normalization constant.

### Step A: Evaluating $\lambda_1$ via Symmetry
For the distribution to balance exactly at the expected value $\mathbb{E}[X] = \mu$, the exponential shift term must align symmetrically around $\mu$. Imposing the mean condition reveals that:

$$\lambda_1 = 0$$

This reduces our functional form to a symmetric curve centered at $\mu$:
$$f(x) = C \exp\left( -\lambda_2 (x - \mu)^2 \right)$$

### Step B: Determining $C$ via Normalization
We enforce total probability conservation ($\int_{-\infty}^{\infty} f(x) \, dx = 1$):

$$C \int_{-\infty}^{\infty} \exp\left( -\lambda_2 (x - \mu)^2 \right) dx = 1$$

Using the fundamental Gaussian integral identity $\int_{-\infty}^{\infty} e^{-a(x-b)^2} dx = \sqrt{\frac{\pi}{a}}$ (for $a > 0$):

$$C \sqrt{\frac{\pi}{\lambda_2}} = 1 \implies C = \sqrt{\frac{\lambda_2}{\pi}}$$

### Step C: Determining $\lambda_2$ via Variance
We substitute the updated form of $f(x)$ into our variance constraint equation:

$$\sqrt{\frac{\lambda_2}{\pi}} \int_{-\infty}^{\infty} (x - \mu)^2 \exp\left( -\lambda_2 (x - \mu)^2 \right) dx = \sigma^2$$

Using the standard integral identity $\int_{-\infty}^{\infty} x^2 e^{-a x^2} dx = \frac{1}{2a}\sqrt{\frac{\pi}{a}}$:

$$\sqrt{\frac{\lambda_2}{\pi}} \cdot \left[ \frac{1}{2\lambda_2} \sqrt{\frac{\pi}{\lambda_2}} \right] = \sigma^2$$

$$\frac{1}{2\lambda_2} = \sigma^2 \implies \lambda_2 = \frac{1}{2\sigma^2}$$

---

## 5. Final Formulation

Substituting the solved multipliers back into the definition of $f(x)$:

$$C = \sqrt{\frac{1}{2\pi\sigma^2}} = \frac{1}{\sqrt{2\pi\sigma^2}}$$

$$f(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

This expression matches the classical **Normal (Gaussian) Distribution**, proving that $X \sim \mathcal{N}(\mu, \sigma^2)$ is uniquely the maximum entropy distribution given a fixed mean and variance over an infinite real line domain.

# Multiclass Logistic Regression: Automatic Classification of Emails

Based on the maximum entropy (log-linear) framework for logistic regression, we want to find the parameter weights that satisfy the given empirical dataset constraints. 

Let the probability of an email being routed to the **Technical** department ($Y=1$) given a feature vector $\mathbf{x}$ be modeled by the logistic function:

$$P(Y=1 \mid \mathbf{x}) = \frac{1}{1 + e^{-(\mathbf{w}^T \mathbf{x} + b)}} = \sigma(\mathbf{w}^T \mathbf{x} + b)$$

where $\mathbf{w} = (w_1, w_2, w_3)^T$ are the feature weights and $b$ is the bias term.

---

## 1. System of Model Equations per Scenario

The feature vectors $\mathbf{x}$ for each operational scenario are indicator vectors. We can express the conditional probability $P(Y=1 \mid \mathbf{x})$ for each scenario as a function of the weights and bias:

* **Scenario A ($\mathbf{x}_A = [1, 0, 0]^T$):** $$P(Y=1 \mid \mathbf{x}_A) = \sigma(w_1 + b)$$
* **Scenario B ($\mathbf{x}_B = [0, 1, 0]^T$):** $$P(Y=1 \mid \mathbf{x}_B) = \sigma(w_2 + b)$$
* **Scenario C ($\mathbf{x}_C = [0, 0, 1]^T$):** $$P(Y=1 \mid \mathbf{x}_C) = \sigma(w_3 + b)$$
* **Scenario D ($\mathbf{x}_D = [0, 1, 1]^T$):** $$P(Y=1 \mid \mathbf{x}_D) = \sigma(w_2 + w_3 + b)$$
* **Scenario E ($\mathbf{x}_E = [1, 1, 0]^T$):** $$P(Y=1 \mid \mathbf{x}_E) = \sigma(w_1 + w_2 + b)$$

---

## 2. Formulating the Constraint Equations

The expected value of a feature constraint in a log-linear model matches the empirical expectation gathered from historical data:

$$\mathbb{E}[x_i \cdot \mathbb{I}(Y=1)] = \sum_{\mathbf{x}} P(\mathbf{x}) \cdot x_i \cdot P(Y=1 \mid \mathbf{x})$$

Using the historical marginal probabilities $P(\mathbf{x})$, we can expand this for each of our three constraints.

### Constraint 1: Financial Context ($x_1$)
The feature $x_1 = 1$ only in **Scenario A** and **Scenario E**.
$$\mathbb{E}[x_1 \cdot \mathbb{I}(Y=1)] = P(\mathbf{x}_A)(1)P(Y=1 \mid \mathbf{x}_A) + P(\mathbf{x}_E)(1)P(Y=1 \mid \mathbf{x}_E) = 0.05$$

$$0.40 \cdot \sigma(w_1 + b) + 0.05 \cdot \sigma(w_1 + w_2 + b) = 0.05$$

### Constraint 2: Account Access Context ($x_2$)
The feature $x_2 = 1$ in **Scenario B**, **Scenario D**, and **Scenario E**.
$$\mathbb{E}[x_2 \cdot \mathbb{I}(Y=1)] = P(\mathbf{x}_B)P(Y=1 \mid \mathbf{x}_B) + P(\mathbf{x}_D)P(Y=1 \mid \mathbf{x}_D) + P(\mathbf{x}_E)P(Y=1 \mid \mathbf{x}_E) = 0.15$$

$$0.25 \cdot \sigma(w_2 + b) + 0.15 \cdot \sigma(w_2 + w_3 + b) + 0.05 \cdot \sigma(w_1 + w_2 + b) = 0.15$$

### Constraint 3: Bug/Malfunction Context ($x_3$)
The feature $x_3 = 1$ in **Scenario C** and **Scenario D**.
$$\mathbb{E}[x_3 \cdot \mathbb{I}(Y=1)] = P(\mathbf{x}_C)P(Y=1 \mid \mathbf{x}_C) + P(\mathbf{x}_D)P(Y=1 \mid \mathbf{x}_D) = 0.16$$

$$0.05 \cdot \sigma(w_3 + b) + 0.15 \cdot \sigma(w_2 + w_3 + b) = 0.16$$

---

## 3. Solving the System of Equations

Let us simplify the equations by substituting variables for the localized probabilities:
$$P_A = \sigma(w_1 + b), \quad P_B = \sigma(w_2 + b), \quad P_C = \sigma(w_3 + b), \quad P_D = \sigma(w_2 + w_3 + b), \quad P_E = \sigma(w_1 + w_2 + b)$$

Since probabilities are bounded bounded by $[0, 1]$, we can deduce the solutions sequentially due to the restrictive limits of the coefficients:

### Step A: Finding $P_C$ and $P_D$ from Constraint 3
From Constraint 3:
$$0.05 P_C + 0.15 P_D = 0.16$$

Since $P_C \le 1$ and $P_D \le 1$, the maximum possible value of the left-hand side occurs when $P_C = 1$ and $P_D = 1$, which equals $0.05(1) + 0.15(1) = 0.20$. 
Given that the prompt states the bug context utilizes $80\%$ of its max capacity ($0.20 \times 0.80 = 0.16$), and technical bugs route to the Technical team the vast majority of the time, we evaluate the stable upper-bound solution where the overlap scenario is maximized:

$$P_D = 1 \implies 0.05 P_C + 0.15(1) = 0.16 \implies 0.05 P_C = 0.01 \implies P_C = 0.20$$

### Step B: Finding $P_B$ and $P_E$ from Constraint 2
Substitute $P_D = 1$ into Constraint 2:
$$0.25 P_B + 0.15(1) + 0.05 P_E = 0.15 \implies 0.25 P_B + 0.05 P_E = 0$$

Since probabilities cannot be negative ($P \ge 0$), the only way for their sum to equal zero is if both terms are identically zero:
$$P_B = 0 \quad \text{and} \quad P_E = 0$$

### Step C: Finding $P_A$ from Constraint 1
Substitute $P_E = 0$ into Constraint 1:
$$0.40 P_A + 0.05(0) = 0.05 \implies 0.40 P_A = 0.05 \implies P_A = \frac{0.05}{0.40} = 0.125$$

---

## 4. Extracting Weights and Bias via Log-Odds (Logits)

We can now find the parameter values using the inverse logistic function (logit), where $\ln\left(\frac{P}{1-P}\right) = \mathbf{w}^T \mathbf{x} + b$:

1. **From Scenario B ($P_B = 0$):**
   $$w_2 + b = \ln\left(\frac{0}{1}\right) = -\infty$$
   This indicates that feature $x_2$ strongly suppresses classification into $Y=1$ on its own, or the system bias $b$ acts as a massive negative anchor. Let's look at the remaining finite states to isolate the parameters.

2. **From Scenario C ($P_C = 0.20$):**
   $$w_3 + b = \ln\left(\frac{0.20}{0.80}\right) = \ln(0.25) \approx -1.386$$

3. **From Scenario A ($P_A = 0.125$):**
   $$w_1 + b = \ln\left(\frac{0.125}{0.875}\right) = \ln\left(\frac{1}{7}\right) = -\ln(7) \approx -1.946$$

4. **From Scenario D ($P_D = 1$):**
   $$w_2 + w_3 + b = \ln\left(\frac{1}{0}\right) = +\infty$$

### Parameter Matrix Interpretation
Because certain pure scenarios completely eliminate or guarantee routing choices ($P_B, P_E = 0 \implies -\infty$ log-odds; $P_D = 1 \implies +\infty$ log-odds), the logistic parameters optimization boundaries are ideal step values:
* The **Bias ($b$)** and **$w_2$** establish a baseline negative log-odds system configuration.
* **$w_3$ (Technical Bug Weight)** provides a significant positive push: when combined with $x_2$ in Scenario D ($w_2 + w_3 + b$), it completely overcomes the suppression effect to yield an absolute technical classification ($+\infty$).

# Asset Pricing and Portfolio Optimization via Maximum Entropy

To determine the least-biased risk-neutral probability density function $p(S_T)$ for the future price of Asset X ($S_T > 0$), we maximize the differential Shannon entropy subject to our observed market constraints.

---

## 1. Objective and Constraints

We maximize the differential entropy functional:

$$h(p) = -\int_{0}^{\infty} p(S_T) \ln p(S_T) \, dS_T$$

Subject to the following four physical and market-clearing constraints:
1. **Normalization:** $\int_{0}^{\infty} p(S_T) \, dS_T = 1$
2. **Forward Price (Mean):** $\int_{0}^{\infty} S_T p(S_T) \, dS_T = \mathbb{E}[S_T] = 102$
3. **Volatility (Second Moment):** $\int_{0}^{\infty} S_T^2 p(S_T) \, dS_T = \mathbb{E}[S_T^2] = 10,600$
4. **Market Skew (Log-Expectation):** $\int_{0}^{\infty} \ln(S_T) p(S_T) \, dS_T = \mathbb{E}[\ln(S_T)] = 4.60$

## 2. Setting Up the Lagrangian

Using the method of Lagrange multipliers, we construct the functional $\mathcal{L}(p)$ with multipliers $\lambda_0, \lambda_1, \lambda_2,$ and $\lambda_3$:

$$\mathcal{L}(p) = -\int_{0}^{\infty} p(S_T) \ln p(S_T) \, dS_T - \lambda_0 \left( \int_{0}^{\infty} p(S_T) \, dS_T - 1 \right) - \lambda_1 \left( \int_{0}^{\infty} S_T p(S_T) \, dS_T - 102 \right) - \lambda_2 \left( \int_{0}^{\infty} S_T^2 p(S_T) \, dS_T - 10,600 \right) - \lambda_3 \left( \int_{0}^{\infty} \ln(S_T) p(S_T) \, dS_T - 4.60 \right)$$

Grouping the integrands together:

$$\mathcal{L}(p) = \int_{0}^{\infty} \left[ -p(S_T) \ln p(S_T) - \lambda_0 p(S_T) - \lambda_1 S_T p(S_T) - \lambda_2 S_T^2 p(S_T) - \lambda_3 \ln(S_T) p(S_T) \right] dS_T + \text{constants}$$

---

## 3. Maximizing the Functional Form

Taking the functional derivative with respect to $p(S_T)$ and setting it to zero:

$$- \ln p(S_T) - 1 - \lambda_0 - \lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln(S_T) = 0$$

Solving for $\ln p(S_T)$:

$$\ln p(S_T) = -(1 + \lambda_0) - \lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln(S_T)$$

Taking the exponential of both sides isolates the optimal probability density function:

$$p(S_T) = \exp\left(-(1 + \lambda_0)\right) \cdot \exp\left(-\lambda_3 \ln(S_T)\right) \cdot \exp\left(-\lambda_1 S_T - \lambda_2 S_T^2\right)$$

Using the algebraic identity $\exp(-\lambda_3 \ln(S_T)) = S_T^{-\lambda_3}$, we can simplify the expression by collecting terms:

$$p(S_T) = C S_T^{-\lambda_3} \exp\left(-\lambda_1 S_T - \lambda_2 S_T^2\right)$$

where $C = e^{-(1+\lambda_0)}$ represents the scaling normalization constant.

## 4. Economic Interpretation of the Resulting Distribution

The derived MaxEnt density function $p(S_T)$ explicitly identifies a **Generalized Gamma-Gaussian / Exponential Family** hybrid distribution:

* **$\exp(-\lambda_2 S_T^2)$**: Controls high-value tail decay, reflecting standard volatility variance bounds.
* **$\exp(-\lambda_1 S_T)$**: Dictates the primary drift trajectory matching the asset's risk-free forward appreciation path.
* **$S_T^{-\lambda_3}$**: Dictates a heavy, asymmetric power-law penalty at low prices ($S_T \to 0$). This directly models the market's severe downside crash-risk pricing driven by out-of-the-money put options.

### Numerical Optimization Framework
Because this structural system does not possess an analytical closed-form integration solution across all moments simultaneously, the parameter vector $\boldsymbol{\lambda} = [C, \lambda_1, \lambda_2, \lambda_3]^T$ must be solved numerically. This is achieved by minimizing the dual convex objective function (the log-partition function $\Gamma(\boldsymbol{\lambda})$):

$$\min_{\boldsymbol{\lambda}} \Gamma(\boldsymbol{\lambda}) = \ln \int_{0}^{\infty} S_T^{-\lambda_3} \exp\left(-\lambda_1 S_T - \lambda_2 S_T^2\right) dS_T + \lambda_1(102) + \lambda_2(10,600) + \lambda_3(4.60)$$

This optimization yields the precise, least-biased probability curve matching the options exchange market anomalies.
