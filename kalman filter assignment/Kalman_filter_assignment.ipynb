# Q. Analytical Derivation

Consider the filter model
\begin{align*}
x^{-}_k &= A_{k-1}\,x^{+}_{k-1} + G_{k-1}\,w_{k-1}, \\
y^{-}_k &= H_k\,x^{-}_k + z_k,
\end{align*}
where
$$x^{+}_{k-1} \sim \mathscr{N}(m_{k-1}, P_{k-1}),$$

Answer the followings:

1. Show that
$$x_k^- \sim \mathscr{N}(m_k^-, P_k^-),$$
where
$$m_k^- = A_{k-1}m_{k-1},$$
and
$$
P_k^- = A_{k-1}P_{k-1}A_{k-1}^T + G_{k-1}\Sigma_p G_{k-1}^T.
$$

2. Show that
$$y_k^- \sim \mathscr{N}(H_k m_k^-, H_k P_k^- H_k^T + \Sigma_m).$$

3. Show that
$$\begin{bmatrix} x_k^- \\ y^{-}_k \end{bmatrix} \sim \mathscr{N}\left(
\begin{bmatrix} m_k^- \\ H_k m_k^- \end{bmatrix},
\begin{bmatrix} P_k^- & P_k^- H_k^T \\ H_k P_k^- & H_k P_k^- H_k^T + \Sigma_m \end{bmatrix}
\right).$$

4. Show that
$$x^{+}_k\triangleq (x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}) \sim \mathscr{N}(m_k, P_k),$$
where the updated mean $m_k$ and updated covariance $P_k$ are given by:
\begin{align*}
K_k &\triangleq P_k^- H_k^T (H_k P_k^- H_k^T + \Sigma_m)^{-1}, \\
m_k &= m_k^- + K_k (y^{\mathrm{obs}}_{k} - H_k m_k^-), \\
P_k &= (I - K_k H_k) P_k^-.
\end{align*}

5. Find
$\mathbb{E}[x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}]$  and
$\text{Var}(x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k})$.

### 1. State Prediction Profile

Since $x_k^-$ is a linear combination of independent Gaussian random variables ($x^{+}_{k-1}$ and $w_{k-1}$), its resulting distribution remains Gaussian. We can define this profile entirely by evaluating its expected value and covariance matrix.

**Mean Vector ($m_k^-$):**
\begin{align*}
m_k^- &= \mathbb{E}[x_k^-] \\
&= \mathbb{E}[A_{k-1}x^{+}_{k-1} + G_{k-1}w_{k-1}] \\
&= A_{k-1}\mathbb{E}[x^{+}_{k-1}] + G_{k-1}\mathbb{E}[w_{k-1}]
\end{align*}

Given that $\mathbb{E}[x^{+}_{k-1}] = m_{k-1}$ and the process noise zero-mean property $\mathbb{E}[w_{k-1}] = 0$:
$$m_k^- = A_{k-1}m_{k-1}$$

**Covariance Matrix ($P_k^-$):**
\begin{align*}
P_k^- &= \mathbb{E}\left[(x_k^- - m_k^-)(x_k^- - m_k^-)^T\right] \\
&= \mathbb{E}\left[\Big(A_{k-1}(x^{+}_{k-1} - m_{k-1}) + G_{k-1}w_{k-1}\Big)\Big(A_{k-1}(x^{+}_{k-1} - m_{k-1}) + G_{k-1}w_{k-1}\Big)^T\right]
\end{align*}

Expanding this expression and applying the fact that state variables and future noise inputs are mutually independent (causing cross-covariance blocks to equal zero):
\begin{align*}
P_k^- &= A_{k-1}\mathbb{E}\left[(x^{+}_{k-1} - m_{k-1})(x^{+}_{k-1} - m_{k-1})^T\right]A_{k-1}^T + G_{k-1}\mathbb{E}[w_{k-1}w_{k-1}^T]G_{k-1}^T \\
&= A_{k-1}P_{k-1}A_{k-1}^T + G_{k-1}\Sigma_p G_{k-1}^T
\end{align*}

Hence, $x_k^- \sim \mathscr{N}(m_k^-, P_k^-)$.

---

### 2. Measurement Prediction Profile

The measurement vector $y_k^-$ is a linear mapping of the predicted Gaussian state vector added to an independent Gaussian measurement noise parameter.

**Mean Vector:**
\begin{align*}
\mathbb{E}[y_k^-] &= \mathbb{E}[H_k x_k^- + z_k] \\
&= H_k \mathbb{E}[x_k^-] + \mathbb{E}[z_k]
\end{align*}

Substituting $\mathbb{E}[x_k^-] = m_k^-$ and $\mathbb{E}[z_k] = 0$:
$$\mathbb{E}[y_k^-] = H_k m_k^-$$

**Covariance Matrix:**
\begin{align*}
\text{Var}(y_k^-) &= \mathbb{E}\left[(y_k^- - H_k m_k^-)(y_k^- - H_k m_k^-)^T\right] \\
&= \mathbb{E}\left[\Big(H_k(x_k^- - m_k^-) + z_k\Big)\Big(H_k(x_k^- - m_k^-) + z_k\Big)^T\right]
\end{align*}

Since the state prediction deviations and tracking errors are uncorrelated, the cross-product expectations drop to zero:
\begin{align*}
\text{Var}(y_k^-) &= H_k \mathbb{E}\left[(x_k^- - m_k^-)(x_k^- - m_k^-)^T\right] H_k^T + \mathbb{E}[z_k z_k^T] \\
&= H_k P_k^- H_k^T + \Sigma_m
\end{align*}

Hence, $y_k^- \sim \mathscr{N}(H_k m_k^-, H_k P_k^- H_k^T + \Sigma_m)$.

---

### 3. Joint Distribution Construction

We can bundle the variables into a combined block matrix form. Given that both vectors are individually Gaussian and linearly related, their joint profile is also joint Gaussian. The cross-covariance components are derived as:

\begin{align*}
\text{Cov}(x_k^-, y_k^-) &= \mathbb{E}\left[(x_k^- - m_k^-)(y_k^- - H_k m_k^-)^T\right] \\
&= \mathbb{E}\left[(x_k^- - m_k^-)\left(H_k(x_k^- - m_k^-) + z_k\right)^T\right] \\
&= \mathbb{E}\left[(x_k^- - m_k^-)(x_k^- - m_k^-)^T\right]H_k^T + \mathbb{E}\left[(x_k^- - m_k^-)z_k^T\right]
\end{align*}

Because the predicted state is uncorrelated with current observation noise, the term $\mathbb{E}\left[(x_k^- - m_k^-)z_k^T\right] = 0$, giving:
$$\text{Cov}(x_k^-, y_k^-) = P_k^- H_k^T$$

By structural symmetry, the bottom-left cross block is simply the transpose:
$$\text{Cov}(y_k^-, x_k^-) = \text{Cov}(x_k^-, y_k^-)^T = H_k P_k^-$$

Arranging these components into the complete joint block structure:
$$\begin{bmatrix} x_k^- \\ y^{-}_k \end{bmatrix} \sim \mathscr{N}\left(
\begin{bmatrix} m_k^- \\ H_k m_k^- \end{bmatrix},
\begin{bmatrix} P_k^- & P_k^- H_k^T \\ H_k P_k^- & H_k P_k^- H_k^T + \Sigma_m \end{bmatrix}
\right)$$

---

### 4. Measurement Update Conditioning

By applying the **Linear Gaussian Conditioning Theorem**, for any joint distribution setup defined as:
$$\begin{bmatrix} X \\ Y \end{bmatrix} \sim \mathscr{N}\left( \begin{bmatrix} \mu_x \\ \mu_y \end{bmatrix}, \begin{bmatrix} \Sigma_{xx} & \Sigma_{xy} \\ \Sigma_{yx} & \Sigma_{yy} \end{bmatrix} \right)$$

The posterior distribution given a known value $(X \mid Y = y)$ remains normal with properties:
$$\mathbb{E}[X \mid Y = y] = \mu_x + \Sigma_{xy} \Sigma_{yy}^{-1} (y - \mu_y)$$
$$\text{Var}(X \mid Y = y) = \Sigma_{xx} - \Sigma_{xy} \Sigma_{yy}^{-1} \Sigma_{yx}$$

Mapping our target terms directly into this statistical identity framework:
* $\mu_x = m_k^-$
* $\mu_y = H_k m_k^-$
* $y = y^{\mathrm{obs}}_{k}$
* $\Sigma_{xx} = P_k^-$
* $\Sigma_{xy} = P_k^- H_k^T$
* $\Sigma_{yx} = H_k P_k^-$
* $\Sigma_{yy} = H_k P_k^- H_k^T + \Sigma_m$

**Kalman Gain definition ($K_k$):**
$$K_k \triangleq \Sigma_{xy} \Sigma_{yy}^{-1} = P_k^- H_k^T (H_k P_k^- H_k^T + \Sigma_m)^{-1}$$

**Updated Mean Evaluation ($m_k$):**
$$m_k = \mu_x + K_k (y^{\mathrm{obs}}_{k} - \mu_y) = m_k^- + K_k (y^{\mathrm{obs}}_{k} - H_k m_k^-)$$

**Updated Covariance Evaluation ($P_k$):**
$$P_k = \Sigma_{xx} - K_k \Sigma_{yx} = P_k^- - K_k (H_k P_k^-) = (I - K_k H_k) P_k^-$$

Hence, $x^{+}_k\triangleq (x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}) \sim \mathscr{N}(m_k, P_k)$.

---

### 5. Conditional Expectation and Variance Summary

Evaluating the exact structural definitions established in **Step 4** directly provides our conditional properties:

$$\mathbb{E}[x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}] = m_k^- + K_k (y^{\mathrm{obs}}_{k} - H_k m_k^-)$$

$$\text{Var}(x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}) = (I - K_k H_k) P_k^-$$

where the optimal weighting gain matrix matches:
$$K_k = P_k^- H_k^T (H_k P_k^- H_k^T + \Sigma_m)^{-1}$$

# Q. 1-D Example

Consider the scalar linear-Gaussian filter model
\begin{aligned}
x^-_k &= a\,x^+_{k-1} + w_{k-1},\qquad w_{k-1}\sim\mathscr N(0,q),\\
y^-_k &= h\,x^-_k + z_k,\qquad\;\;\;\; z_k\sim\mathscr N(0,r),
\end{aligned}
with prior $x_0\sim \mathscr{N}(m_0,P_0)$. Define $Y_{k}=\{y^{\mathrm{obs}}_1,\dots,y^{\mathrm{obs}}_{k}\}$.

Answer the following:

1. Show that
\begin{aligned}
m_k^- &= a\,m_{k-1},\\
P_k^- &= a^2 P_{k-1} + q.
\end{aligned}

2. Show that
\begin{aligned}
m_k &= m_k^- + K_k\,v_k
= m_k^- + \frac{P_k^- h}{S_k}\bigl(y^{\mathrm{obs}}_k - h\,m_k^-\bigr),\\
P_k &= (1 - K_k h)\,P_k^-
= \Bigl(1 - \frac{P_k^- h^2}{S_k}\Bigr) P_k^- .
\end{aligned}

3. Show that the predictive measurement distribution (before seeing $y_k$):
\begin{align*}
p(y^-_k | Y_{k-1})=\mathscr N\bigl(h\,m_k^-,\; h^2 P_k^- + r\bigr).
\end{align*}

4. Show that the posterior-predictive measurement distribution (after filtering on $y_k$):
\begin{align*}
p(y^-_k\mid Y_k)=\mathscr N\bigl(h\,m_k,\; h^2 P_k + r\bigr).
\end{align*}

5. For suitable numerical values for $a,q,h,r$ animate the above to prior distribution and the posterior distribution.

### 1. State Prediction Setup

Since the system models a scalar process, we can find the mean and variance directly using expectation rules for scalar random variables:

**Predicted Mean ($m_k^-$):**
\begin{align*}
m_k^- &= \mathbb{E}[x^-_k \mid Y_{k-1}] \\
&= \mathbb{E}[a\,x^+_{k-1} + w_{k-1} \mid Y_{k-1}] \\
&= a\,\mathbb{E}[x^+_{k-1} \mid Y_{k-1}] + \mathbb{E}[w_{k-1}]
\end{align*}

Using the prior mean tracking definition $\mathbb{E}[x^+_{k-1} \mid Y_{k-1}] = m_{k-1}$ and the zero-mean noise property $\mathbb{E}[w_{k-1}] = 0$:
$$m_k^- = a\,m_{k-1}$$

**Predicted Variance ($P_k^-$):**
\begin{align*}
P_k^- &= \text{Var}(x^-_k \mid Y_{k-1}) \\
&= \text{Var}(a\,x^+_{k-1} + w_{k-1} \mid Y_{k-1})
\end{align*}

Because the current state $x^+_{k-1}$ and the next process disturbance $w_{k-1}$ are independent, their variances add directly, and the scalar multiplier squares when factored out:
\begin{align*}
P_k^- &= a^2 \text{Var}(x^+_{k-1} \mid Y_{k-1}) + \text{Var}(w_{k-1}) \\
&= a^2 P_{k-1} + q
\end{align*}

---

### 2. Measurement Update and Kalman Gain

By utilizing the standard conditioning rules for joint Gaussian variables, the updated scalar state mean and variance are obtained as follows:

**Innovation Variance definition ($S_k$):**
$$S_k = \text{Var}(y^-_k \mid Y_{k-1}) = h^2 P_k^- + r$$

**Kalman Gain ($K_k$):**
$$K_k = \frac{\text{Cov}(x^-_k, y^-_k \mid Y_{k-1})}{\text{Var}(y^-_k \mid Y_{k-1})} = \frac{P_k^- h}{S_k} = \frac{P_k^- h}{h^2 P_k^- + r}$$

**Updated Mean ($m_k$):**
Using the innovation residual $v_k = y^{\mathrm{obs}}_k - h\,m_k^-$:
$$m_k = m_k^- + K_k\,v_k = m_k^- + \frac{P_k^- h}{S_k}\bigl(y^{\mathrm{obs}}_k - h\,m_k^-\bigr)$$

**Updated Variance ($P_k$):**
\begin{align*}
P_k &= P_k^- - K_k^2 S_k \\
&= P_k^- - K_k \left(\frac{P_k^- h}{S_k}\right) S_k \\
&= P_k^- - K_k h P_k^- \\
&= (1 - K_k h)\,P_k^-
\end{align*}

Substituting the structural definition of $K_k$:
$$P_k = \Bigl(1 - \frac{P_k^- h^2}{S_k}\Bigr) P_k^-$$

---

### 3. Prior Predictive Measurement Distribution

Before processing the current observed sample $y^{\mathrm{obs}}_k$, the distribution of $y^-_k$ relies entirely on the predicted state information $x^-_k \mid Y_{k-1} \sim \mathscr{N}(m_k^-, P_k^-)$.

**Expected Measurement:**
$$\mathbb{E}[y^-_k \mid Y_{k-1}] = \mathbb{E}[h\,x^-_k + z_k \mid Y_{k-1}] = h\,\mathbb{E}[x^-_k \mid Y_{k-1}] + \mathbb{E}[z_k] = h\,m_k^-$$

**Measurement Variance:**
Since the state prediction deviations and sensor noise are statistically independent:
$$\text{Var}(y^-_k \mid Y_{k-1}) = \text{Var}(h\,x^-_k + z_k \mid Y_{k-1}) = h^2 \text{Var}(x^-_k \mid Y_{k-1}) + \text{Var}(z_k) = h^2 P_k^- + r$$

Hence, the predictive distribution simplifies directly to:
$$p(y^-_k \mid Y_{k-1}) = \mathscr{N}\bigl(h\,m_k^-,\; h^2 P_k^- + r\bigr)$$

---

### 4. Posterior Predictive Measurement Distribution

After assimilation of the current measurement, our belief about the underlying state transitions to the filtered profile $x_k \mid Y_k \sim \mathscr{N}(m_k, P_k)$. A hypothetical measurement generated at this stage yields:

**Expected Value:**
$$\mathbb{E}[y^-_k \mid Y_k] = \mathbb{E}[h\,x_k + z_k \mid Y_k] = h\,\mathbb{E}[x_k \mid Y_k] + \mathbb{E}[z_k] = h\,m_k$$

**Variance Value:**
$$\text{Var}(y^-_k \mid Y_k) = \text{Var}(h\,x_k + z_k \mid Y_k) = h^2 \text{Var}(x_k \mid Y_k) + \text{Var}(z_k) = h^2 P_k + r$$

Hence, the distribution follows:
$$p(y^-_k \mid Y_k) = \mathscr{N}\bigl(h\,m_k,\; h^2 P_k + r\bigr)$$
