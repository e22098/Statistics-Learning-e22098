# Q. Analytical Derivation

Consider the filter model
\begin{align*}
x^{-}_k &= A_{k-1}\,x^{+}_{k-1} + G_{k-1}\,w_{k-1}, \\
y^{-}_k &= H_k\,x^{-}_k + z_k,
\end{align*}
where
$$x^{+}_{k-1} \sim \mathscr{N}(m_{k-1}, P_{k-1}),$$

Answer the followings:

1. Show that
$$x_k^- \sim \mathscr{N}(m_k^-, P_k^-),$$
where
$$m_k^- = A_{k-1}m_{k-1},$$
and
$$
P_k^- = A_{k-1}P_{k-1}A_{k-1}^T + G_{k-1}\Sigma_p G_{k-1}^T.
$$

2. Show that
$$y_k^- \sim \mathscr{N}(H_k m_k^-, H_k P_k^- H_k^T + \Sigma_m).$$

3. Show that
$$\begin{bmatrix} x_k^- \\ y^{-}_k \end{bmatrix} \sim \mathscr{N}\left(
\begin{bmatrix} m_k^- \\ H_k m_k^- \end{bmatrix},
\begin{bmatrix} P_k^- & P_k^- H_k^T \\ H_k P_k^- & H_k P_k^- H_k^T + \Sigma_m \end{bmatrix}
\right).$$

4. Show that
$$x^{+}_k\triangleq (x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}) \sim \mathscr{N}(m_k, P_k),$$
where the updated mean $m_k$ and updated covariance $P_k$ are given by:
\begin{align*}
K_k &\triangleq P_k^- H_k^T (H_k P_k^- H_k^T + \Sigma_m)^{-1}, \\
m_k &= m_k^- + K_k (y^{\mathrm{obs}}_{k} - H_k m_k^-), \\
P_k &= (I - K_k H_k) P_k^-.
\end{align*}

5. Find
$\mathbb{E}[x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}]$  and
$\text{Var}(x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k})$.

### 1. State Prediction Profile

Since $x_k^-$ is a linear combination of independent Gaussian random variables ($x^{+}_{k-1}$ and $w_{k-1}$), its resulting distribution remains Gaussian. We can define this profile entirely by evaluating its expected value and covariance matrix.

**Mean Vector ($m_k^-$):**
\begin{align*}
m_k^- &= \mathbb{E}[x_k^-] \\
&= \mathbb{E}[A_{k-1}x^{+}_{k-1} + G_{k-1}w_{k-1}] \\
&= A_{k-1}\mathbb{E}[x^{+}_{k-1}] + G_{k-1}\mathbb{E}[w_{k-1}]
\end{align*}

Given that $\mathbb{E}[x^{+}_{k-1}] = m_{k-1}$ and the process noise zero-mean property $\mathbb{E}[w_{k-1}] = 0$:
$$m_k^- = A_{k-1}m_{k-1}$$

**Covariance Matrix ($P_k^-$):**
\begin{align*}
P_k^- &= \mathbb{E}\left[(x_k^- - m_k^-)(x_k^- - m_k^-)^T\right] \\
&= \mathbb{E}\left[\Big(A_{k-1}(x^{+}_{k-1} - m_{k-1}) + G_{k-1}w_{k-1}\Big)\Big(A_{k-1}(x^{+}_{k-1} - m_{k-1}) + G_{k-1}w_{k-1}\Big)^T\right]
\end{align*}

Expanding this expression and applying the fact that state variables and future noise inputs are mutually independent (causing cross-covariance blocks to equal zero):
\begin{align*}
P_k^- &= A_{k-1}\mathbb{E}\left[(x^{+}_{k-1} - m_{k-1})(x^{+}_{k-1} - m_{k-1})^T\right]A_{k-1}^T + G_{k-1}\mathbb{E}[w_{k-1}w_{k-1}^T]G_{k-1}^T \\
&= A_{k-1}P_{k-1}A_{k-1}^T + G_{k-1}\Sigma_p G_{k-1}^T
\end{align*}

Hence, $x_k^- \sim \mathscr{N}(m_k^-, P_k^-)$.

---

### 2. Measurement Prediction Profile

The measurement vector $y_k^-$ is a linear mapping of the predicted Gaussian state vector added to an independent Gaussian measurement noise parameter.

**Mean Vector:**
\begin{align*}
\mathbb{E}[y_k^-] &= \mathbb{E}[H_k x_k^- + z_k] \\
&= H_k \mathbb{E}[x_k^-] + \mathbb{E}[z_k]
\end{align*}

Substituting $\mathbb{E}[x_k^-] = m_k^-$ and $\mathbb{E}[z_k] = 0$:
$$\mathbb{E}[y_k^-] = H_k m_k^-$$

**Covariance Matrix:**
\begin{align*}
\text{Var}(y_k^-) &= \mathbb{E}\left[(y_k^- - H_k m_k^-)(y_k^- - H_k m_k^-)^T\right] \\
&= \mathbb{E}\left[\Big(H_k(x_k^- - m_k^-) + z_k\Big)\Big(H_k(x_k^- - m_k^-) + z_k\Big)^T\right]
\end{align*}

Since the state prediction deviations and tracking errors are uncorrelated, the cross-product expectations drop to zero:
\begin{align*}
\text{Var}(y_k^-) &= H_k \mathbb{E}\left[(x_k^- - m_k^-)(x_k^- - m_k^-)^T\right] H_k^T + \mathbb{E}[z_k z_k^T] \\
&= H_k P_k^- H_k^T + \Sigma_m
\end{align*}

Hence, $y_k^- \sim \mathscr{N}(H_k m_k^-, H_k P_k^- H_k^T + \Sigma_m)$.

---

### 3. Joint Distribution Construction

We can bundle the variables into a combined block matrix form. Given that both vectors are individually Gaussian and linearly related, their joint profile is also joint Gaussian. The cross-covariance components are derived as:

\begin{align*}
\text{Cov}(x_k^-, y_k^-) &= \mathbb{E}\left[(x_k^- - m_k^-)(y_k^- - H_k m_k^-)^T\right] \\
&= \mathbb{E}\left[(x_k^- - m_k^-)\left(H_k(x_k^- - m_k^-) + z_k\right)^T\right] \\
&= \mathbb{E}\left[(x_k^- - m_k^-)(x_k^- - m_k^-)^T\right]H_k^T + \mathbb{E}\left[(x_k^- - m_k^-)z_k^T\right]
\end{align*}

Because the predicted state is uncorrelated with current observation noise, the term $\mathbb{E}\left[(x_k^- - m_k^-)z_k^T\right] = 0$, giving:
$$\text{Cov}(x_k^-, y_k^-) = P_k^- H_k^T$$

By structural symmetry, the bottom-left cross block is simply the transpose:
$$\text{Cov}(y_k^-, x_k^-) = \text{Cov}(x_k^-, y_k^-)^T = H_k P_k^-$$

Arranging these components into the complete joint block structure:
$$\begin{bmatrix} x_k^- \\ y^{-}_k \end{bmatrix} \sim \mathscr{N}\left(
\begin{bmatrix} m_k^- \\ H_k m_k^- \end{bmatrix},
\begin{bmatrix} P_k^- & P_k^- H_k^T \\ H_k P_k^- & H_k P_k^- H_k^T + \Sigma_m \end{bmatrix}
\right)$$

---

### 4. Measurement Update Conditioning

By applying the **Linear Gaussian Conditioning Theorem**, for any joint distribution setup defined as:
$$\begin{bmatrix} X \\ Y \end{bmatrix} \sim \mathscr{N}\left( \begin{bmatrix} \mu_x \\ \mu_y \end{bmatrix}, \begin{bmatrix} \Sigma_{xx} & \Sigma_{xy} \\ \Sigma_{yx} & \Sigma_{yy} \end{bmatrix} \right)$$

The posterior distribution given a known value $(X \mid Y = y)$ remains normal with properties:
$$\mathbb{E}[X \mid Y = y] = \mu_x + \Sigma_{xy} \Sigma_{yy}^{-1} (y - \mu_y)$$
$$\text{Var}(X \mid Y = y) = \Sigma_{xx} - \Sigma_{xy} \Sigma_{yy}^{-1} \Sigma_{yx}$$

Mapping our target terms directly into this statistical identity framework:
* $\mu_x = m_k^-$
* $\mu_y = H_k m_k^-$
* $y = y^{\mathrm{obs}}_{k}$
* $\Sigma_{xx} = P_k^-$
* $\Sigma_{xy} = P_k^- H_k^T$
* $\Sigma_{yx} = H_k P_k^-$
* $\Sigma_{yy} = H_k P_k^- H_k^T + \Sigma_m$

**Kalman Gain definition ($K_k$):**
$$K_k \triangleq \Sigma_{xy} \Sigma_{yy}^{-1} = P_k^- H_k^T (H_k P_k^- H_k^T + \Sigma_m)^{-1}$$

**Updated Mean Evaluation ($m_k$):**
$$m_k = \mu_x + K_k (y^{\mathrm{obs}}_{k} - \mu_y) = m_k^- + K_k (y^{\mathrm{obs}}_{k} - H_k m_k^-)$$

**Updated Covariance Evaluation ($P_k$):**
$$P_k = \Sigma_{xx} - K_k \Sigma_{yx} = P_k^- - K_k (H_k P_k^-) = (I - K_k H_k) P_k^-$$

Hence, $x^{+}_k\triangleq (x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}) \sim \mathscr{N}(m_k, P_k)$.

---

### 5. Conditional Expectation and Variance Summary

Evaluating the exact structural definitions established in **Step 4** directly provides our conditional properties:

$$\mathbb{E}[x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}] = m_k^- + K_k (y^{\mathrm{obs}}_{k} - H_k m_k^-)$$

$$\text{Var}(x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}) = (I - K_k H_k) P_k^-$$

where the optimal weighting gain matrix matches:
$$K_k = P_k^- H_k^T (H_k P_k^- H_k^T + \Sigma_m)^{-1}$$

# Q. 1-D Example

Consider the scalar linear-Gaussian filter model
\begin{aligned}
x^-_k &= a\,x^+_{k-1} + w_{k-1},\qquad w_{k-1}\sim\mathscr N(0,q),\\
y^-_k &= h\,x^-_k + z_k,\qquad\;\;\;\; z_k\sim\mathscr N(0,r),
\end{aligned}
with prior $x_0\sim \mathscr{N}(m_0,P_0)$. Define $Y_{k}=\{y^{\mathrm{obs}}_1,\dots,y^{\mathrm{obs}}_{k}\}$.

Answer the following:

1. Show that
\begin{aligned}
m_k^- &= a\,m_{k-1},\\
P_k^- &= a^2 P_{k-1} + q.
\end{aligned}

2. Show that
\begin{aligned}
m_k &= m_k^- + K_k\,v_k
= m_k^- + \frac{P_k^- h}{S_k}\bigl(y^{\mathrm{obs}}_k - h\,m_k^-\bigr),\\
P_k &= (1 - K_k h)\,P_k^-
= \Bigl(1 - \frac{P_k^- h^2}{S_k}\Bigr) P_k^- .
\end{aligned}

3. Show that the predictive measurement distribution (before seeing $y_k$):
\begin{align*}
p(y^-_k | Y_{k-1})=\mathscr N\bigl(h\,m_k^-,\; h^2 P_k^- + r\bigr).
\end{align*}

4. Show that the posterior-predictive measurement distribution (after filtering on $y_k$):
\begin{align*}
p(y^-_k\mid Y_k)=\mathscr N\bigl(h\,m_k,\; h^2 P_k + r\bigr).
\end{align*}

5. For suitable numerical values for $a,q,h,r$ animate the above to prior distribution and the posterior distribution.

### 1. State Prediction Setup

Since the system models a scalar process, we can find the mean and variance directly using expectation rules for scalar random variables:

**Predicted Mean ($m_k^-$):**
\begin{align*}
m_k^- &= \mathbb{E}[x^-_k \mid Y_{k-1}] \\
&= \mathbb{E}[a\,x^+_{k-1} + w_{k-1} \mid Y_{k-1}] \\
&= a\,\mathbb{E}[x^+_{k-1} \mid Y_{k-1}] + \mathbb{E}[w_{k-1}]
\end{align*}

Using the prior mean tracking definition $\mathbb{E}[x^+_{k-1} \mid Y_{k-1}] = m_{k-1}$ and the zero-mean noise property $\mathbb{E}[w_{k-1}] = 0$:
$$m_k^- = a\,m_{k-1}$$

**Predicted Variance ($P_k^-$):**
\begin{align*}
P_k^- &= \text{Var}(x^-_k \mid Y_{k-1}) \\
&= \text{Var}(a\,x^+_{k-1} + w_{k-1} \mid Y_{k-1})
\end{align*}

Because the current state $x^+_{k-1}$ and the next process disturbance $w_{k-1}$ are independent, their variances add directly, and the scalar multiplier squares when factored out:
\begin{align*}
P_k^- &= a^2 \text{Var}(x^+_{k-1} \mid Y_{k-1}) + \text{Var}(w_{k-1}) \\
&= a^2 P_{k-1} + q
\end{align*}

---

### 2. Measurement Update and Kalman Gain

By utilizing the standard conditioning rules for joint Gaussian variables, the updated scalar state mean and variance are obtained as follows:

**Innovation Variance definition ($S_k$):**
$$S_k = \text{Var}(y^-_k \mid Y_{k-1}) = h^2 P_k^- + r$$

**Kalman Gain ($K_k$):**
$$K_k = \frac{\text{Cov}(x^-_k, y^-_k \mid Y_{k-1})}{\text{Var}(y^-_k \mid Y_{k-1})} = \frac{P_k^- h}{S_k} = \frac{P_k^- h}{h^2 P_k^- + r}$$

**Updated Mean ($m_k$):**
Using the innovation residual $v_k = y^{\mathrm{obs}}_k - h\,m_k^-$:
$$m_k = m_k^- + K_k\,v_k = m_k^- + \frac{P_k^- h}{S_k}\bigl(y^{\mathrm{obs}}_k - h\,m_k^-\bigr)$$

**Updated Variance ($P_k$):**
\begin{align*}
P_k &= P_k^- - K_k^2 S_k \\
&= P_k^- - K_k \left(\frac{P_k^- h}{S_k}\right) S_k \\
&= P_k^- - K_k h P_k^- \\
&= (1 - K_k h)\,P_k^-
\end{align*}

Substituting the structural definition of $K_k$:
$$P_k = \Bigl(1 - \frac{P_k^- h^2}{S_k}\Bigr) P_k^-$$

---

### 3. Prior Predictive Measurement Distribution

Before processing the current observed sample $y^{\mathrm{obs}}_k$, the distribution of $y^-_k$ relies entirely on the predicted state information $x^-_k \mid Y_{k-1} \sim \mathscr{N}(m_k^-, P_k^-)$.

**Expected Measurement:**
$$\mathbb{E}[y^-_k \mid Y_{k-1}] = \mathbb{E}[h\,x^-_k + z_k \mid Y_{k-1}] = h\,\mathbb{E}[x^-_k \mid Y_{k-1}] + \mathbb{E}[z_k] = h\,m_k^-$$

**Measurement Variance:**
Since the state prediction deviations and sensor noise are statistically independent:
$$\text{Var}(y^-_k \mid Y_{k-1}) = \text{Var}(h\,x^-_k + z_k \mid Y_{k-1}) = h^2 \text{Var}(x^-_k \mid Y_{k-1}) + \text{Var}(z_k) = h^2 P_k^- + r$$

Hence, the predictive distribution simplifies directly to:
$$p(y^-_k \mid Y_{k-1}) = \mathscr{N}\bigl(h\,m_k^-,\; h^2 P_k^- + r\bigr)$$

---

### 4. Posterior Predictive Measurement Distribution

After assimilation of the current measurement, our belief about the underlying state transitions to the filtered profile $x_k \mid Y_k \sim \mathscr{N}(m_k, P_k)$. A hypothetical measurement generated at this stage yields:

**Expected Value:**
$$\mathbb{E}[y^-_k \mid Y_k] = \mathbb{E}[h\,x_k + z_k \mid Y_k] = h\,\mathbb{E}[x_k \mid Y_k] + \mathbb{E}[z_k] = h\,m_k$$

**Variance Value:**
$$\text{Var}(y^-_k \mid Y_k) = \text{Var}(h\,x_k + z_k \mid Y_k) = h^2 \text{Var}(x_k \mid Y_k) + \text{Var}(z_k) = h^2 P_k + r$$

Hence, the distribution follows:
$$p(y^-_k \mid Y_k) = \mathscr{N}\bigl(h\,m_k,\; h^2 P_k + r\bigr)$$

### 5. Numerical Simulation and Animation of 1-D Kalman Filter

To animate the dynamic transition between the **Prior Distribution** (state prediction before the measurement), the **Measurement Likelihood**, and the resulting **Posterior Distribution** (after the filtering update), we can implement a numerical simulation in Python.

Below is the complete executable Python script. It utilizes standard parameters ($a=1.0, q=0.5, h=1.0, r=0.8$) and animates how the probability density functions (PDFs) shift and narrow as the measurement vector is assimilated.

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from scipy.stats import norm

# 1. Define Model Parameters
a = 0.95      # State transition multiplier
q = 0.40      # Process noise variance
h = 1.00      # Measurement scaling factor
r = 0.60      # Measurement noise variance

# Initial Prior at k-1
m_prev = 1.0
P_prev = 1.2

# 2. Compute Kalman Filter Step for Time 'k'
# Time Update (Prediction)
m_minus = a * m_prev
P_minus = (a**2) * P_prev + q

# Simulated True Observation
y_obs = 3.5  

# Measurement Update (Correction)
S = (h**2) * P_minus + r
K = (P_minus * h) / S
m_post = m_minus + K * (y_obs - h * m_minus)
P_post = (1 - K * h) * P_minus

# 3. Setup Plotting Canvas
fig, ax = plt.subplots(figsize=(10, 6))
x_domain = np.linspace(-4, 7, 1000)

# Likelihood function envelope (normalized for visualization)
likelihood_pdf = norm.pdf(x_domain, loc=y_obs/h, scale=np.sqrt(r)/h)

ax.plot(x_domain, likelihood_pdf, 'g--', label=f'Measurement Likelihood ($y^{{\mathrm{{obs}}}}_k = {y_obs}$)', alpha=0.6)
ax.fill_between(x_domain, likelihood_pdf, color='green', alpha=0.05)

# Initialize lines for animation tracking
prior_line, = ax.plot([], [], 'r-', lw=2.5, label='Prior Distribution $p(x^-_k \mid Y_{k-1})$')
posterior_line, = ax.plot([], [], 'b-', lw=2.5, label='Posterior Distribution $p(x_k \mid Y_k)$')
prior_fill = ax.fill_between([], [], color='red', alpha=0.1)
posterior_fill = ax.fill_between([], [], color='blue', alpha=0.15)

ax.set_xlim(-4, 7)
ax.set_ylim(0, 1.2)
ax.set_title('1-D Kalman Filter: Dynamic State Distribution Transition', fontsize=12, fontweight='bold')
ax.set_xlabel('State Vector Domain ($x$)', fontsize=10)
ax.set_ylabel('Probability Density', fontsize=10)
ax.grid(True, linestyle=':', alpha=0.6)
ax.legend(loc='upper left')

# 4. Define Animation Core
def init():
    prior_line.set_data([], [])
    posterior_line.set_data([], [])
    return prior_line, posterior_line

def animate(frame):
    global prior_fill, posterior_fill
    # Remove previous fills to prevent ghosting effects
    prior_fill.remove()
    posterior_fill.remove()
    
    # Frame 0 to 40: Fade in Prior
    if frame <= 40:
        alpha_p = frame / 40.0
        alpha_post = 0.0
    # Frame 41 to 80: Morph smoothly into Posterior profile
    elif frame <= 80:
        alpha_p = 1.0
        alpha_post = (frame - 40) / 40.0
    else:
        alpha_p = 1.0
        alpha_post = 1.0

    # Calculate densities
    y_prior = norm.pdf(x_domain, loc=m_minus, scale=np.sqrt(P_minus))
    y_post = norm.pdf(x_domain, loc=m_post, scale=np.sqrt(P_post))
    
    prior_line.set_data(x_domain, y_prior)
    prior_line.set_alpha(alpha_p)
    prior_fill = ax.fill_between(x_domain, y_prior, color='red', alpha=0.1 * alpha_p)
    
    if alpha_post > 0:
        posterior_line.set_data(x_domain, y_post)
        posterior_line.set_alpha(alpha_post)
        posterior_fill = ax.fill_between(x_domain, y_post, color='blue', alpha=0.15 * alpha_post)
    else:
        posterior_line.set_data([], [])
        posterior_fill = ax.fill_between([], [], color='blue', alpha=0)
        
    return prior_line, posterior_line

# 5. Execute Animation Build
ani = animation.FuncAnimation(fig, animate, init_func=init, frames=100, interval=40, blit=True)
plt.show()

# Q. 2D-Position Estimation

## Part A

Consider a two-dimensional constant-velocity dynamical system. The hidden state at time step $k$ is

$$
x_k =
\begin{bmatrix}
p_x(k)\\
p_y(k)\\
v_x(k)\\
v_y(k)
\end{bmatrix},
$$

where $(p_x(k),p_y(k))$ denote the position components and $(v_x(k),v_y(k))$ denote the velocity components.

The measurement consists only of the two position components:

$$
y_k =
\begin{bmatrix}
p_x^{\mathrm{meas}}(k)\\
p_y^{\mathrm{meas}}(k)
\end{bmatrix}.
$$

The linear Gaussian state-space filter model is

$$
x^-_k = A x^+_{k-1} + G w_{k-1},
$$

$$
y^-_k = Hx^+_k + z_k,
$$

where

$$
w_{k-1} \sim \mathscr{N}(0,\Sigma_p),
\qquad
z_k \sim \mathscr{N}(0,\Sigma_m).
$$

The process noise sequence $w_k$, measurement noise sequence $z_k$, and the initial state are assumed mutually independent.

Show that:

$$
A =
\begin{bmatrix}
1 & 0 & \Delta t & 0\\
0 & 1 & 0 & \Delta t\\
0 & 0 & 1 & 0\\
0 & 0 & 0 & 1
\end{bmatrix},
$$



$$
H =
\begin{bmatrix}
1 & 0 & 0 & 0\\
0 & 1 & 0 & 0
\end{bmatrix},
$$

and

$$
G =
\begin{bmatrix}
\frac{1}{2}\Delta t^2 & 0\\
0 & \frac{1}{2}\Delta t^2\\
\Delta t & 0\\
0 & \Delta t
\end{bmatrix}.
$$

### Kinematic Derivation of the State-Space Matrices

To derive the system matrices $A$, $H$, and $G$, we analyze the continuous-time physical relationships governing a constant-velocity kinematic model under an accelerating disturbance, and then map them directly to our state vector $x_k = [p_x(k), p_y(k), v_x(k), v_y(k)]^T$.

---

#### 1. Derivation of State Transition Matrix ($A$) and Noise Coupling Matrix ($G$)

Under a constant-velocity assumption over a small discrete sampling time step $\Delta t$, the position updates based on the current velocity, while the velocity is assumed to remain constant except for the influence of an external random acceleration noise vector $w_{k-1} = [a_x(k-1), a_y(k-1)]^T$.

Applying standard Newtonian equations of motion along the independent $x$ and $y$ orthogonal coordinates yields:

**Position Equations:**
$$p_x(k) = p_x(k-1) + v_x(k-1)\Delta t + \frac{1}{2}a_x(k-1)\Delta t^2$$
$$p_y(k) = p_y(k-1) + v_y(k-1)\Delta t + \frac{1}{2}a_y(k-1)\Delta t^2$$

**Velocity Equations:**
$$v_x(k) = v_x(k-1) + a_x(k-1)\Delta t$$
$$v_y(k) = v_y(k-1) + a_y(k-1)\Delta t$$

We can isolate the deterministic state transitions from the stochastic noise disturbances by structuring these equations into matrix form matching $x^-_k = A x^+_{k-1} + G w_{k-1}$:

$$\begin{bmatrix} p_x(k) \\ p_y(k) \\ v_x(k) \\ v_y(k) \end{bmatrix} = \begin{bmatrix} 1 & 0 & \Delta t & 0 \\ 0 & 1 & 0 & \Delta t \\ 0 & 0 & 1 & 0 \\ 0 & 0 & 0 & 1 \end{bmatrix} \begin{bmatrix} p_x(k-1) \\ p_y(k-1) \\ v_x(k-1) \\ v_y(k-1) \end{bmatrix} + \begin{bmatrix} \frac{1}{2}\Delta t^2 & 0 \\ 0 & \frac{1}{2}\Delta t^2 \\ \Delta t & 0 \\ 0 & \Delta t \end{bmatrix} \begin{bmatrix} a_x(k-1) \\ a_y(k-1) \end{bmatrix}$$

Extracting the respective matrix blocks confirms our definitions:

$$A = \begin{bmatrix} 1 & 0 & \Delta t & 0 \\ 0 & 1 & 0 & \Delta t \\ 0 & 0 & 1 & 0 \\ 0 & 0 & 0 & 1 \end{bmatrix}, \qquad G = \begin{bmatrix} \frac{1}{2}\Delta t^2 & 0 \\ 0 & \frac{1}{2}\Delta t^2 \\ \Delta t & 0 \\ 0 & \Delta t \end{bmatrix}$$

---

#### 2. Derivation of Measurement Matrix ($H$)

The measurement vector $y_k$ isolates only the spatial position coordinates $(p_x, p_y)$ from the full latent tracking state, corrupting them with independent instrumentation noise $z_k$:

$$y_k = \begin{bmatrix} p_x^{\mathrm{meas}}(k) \\ p_y^{\mathrm{meas}}(k) \end{bmatrix} = \begin{bmatrix} p_x(k) \\ p_y(k) \end{bmatrix} + z_k$$

To express this extraction linear mapping in terms of the complete state vector $x_k$, we write:

$$y_k = \begin{bmatrix} 1\cdot p_x(k) + 0\cdot p_y(k) + 0\cdot v_x(k) + 0\cdot v_y(k) \\ 0\cdot p_x(k) + 1\cdot p_y(k) + 0\cdot v_x(k) + 0\cdot v_y(k) \end{bmatrix} + z_k$$

Converting this linear system into its matrix equivalent form $y^-_k = Hx_k + z_k$:

$$y_k = \begin{bmatrix} 1 & 0 & 0 & 0 \\ 0 & 1 & 0 & 0 \end{bmatrix} \begin{bmatrix} p_x(k) \\ p_y(k) \\ v_x(k) \\ v_y(k) \end{bmatrix} + z_k$$

Thus, the measurement matrix is verified as:

$$H = \begin{bmatrix} 1 & 0 & 0 & 0 \\ 0 & 1 & 0 & 0 \end{bmatrix}$$

## Part B

Develop a python based method to filter a sequence of noisy GPS obtained position meaurements.

### Python Implementation of 2D Kalman Filter for Noisy GPS Smoothing

The following script sets up the full constant-velocity state-space model ($A, G, H$), handles the tracking of state estimates and error covariances ($m_k, P_k$), and simulates a noisy flight or driving path to demonstrate the performance of the filter.

```python
import numpy as np
import matplotlib.pyplot as plt

class KalmanFilter2D:
    def __init__(self, dt, q_var, r_var):
        """
        Initializes the 2D Constant-Velocity Kalman Filter.
        
        Parameters:
        dt    : float - Time step increment (Delta t)
        q_var : float - Process noise acceleration variance (sigma_p^2)
        r_var : float - Measurement sensor position variance (sigma_m^2)
        """
        self.dt = dt
        
        # 1. State Transition Matrix (A)
        self.A = np.array([
            [1, 0, dt,  0],
            [0, 1,  0, dt],
            [0, 0,  1,  0],
            [0, 0,  0,  1]
        ])
        
        # 2. Noise Coupling Matrix (G)
        self.G = np.array([
            [0.5 * (dt**2), 0],
            [0, 0.5 * (dt**2)],
            [dt,            0],
            [0,            dt]
        ])
        
        # 3. Measurement Matrix (H)
        self.H = np.array([
            [1, 0, 0, 0],
            [0, 1, 0, 0]
        ])
        
        # 4. Covariance Matrices (Sigma_p and Sigma_m)
        self.Sigma_p = np.eye(2) * q_var
        self.Sigma_m = np.eye(2) * r_var
        
        # Complete Process Noise Matrix Q = G * Sigma_p * G^T
        self.Q = self.G @ self.Sigma_p @ self.G.T
        self.R = self.Sigma_m
        
        # 5. Initialize Latent State Vector [px, py, vx, vy]^T and Covariance P
        self.m = np.array([0.0, 0.0, 0.0, 0.0])
        self.P = np.eye(4) * 10.0  # High initial uncertainty
        
    def initialize_state(self, initial_x, initial_y, initial_vx=0.0, initial_vy=0.0):
        """Seeds the filter state with the first valid observation."""
        self.m = np.array([initial_x, initial_y, initial_vx, initial_vy])
        self.P = np.eye(4) * 1.0

    def predict(self):
        """Time Update Step: Projects the current state estimate forward."""
        self.m = self.A @ self.m
        self.P = (self.A @ self.P @ self.A.T) + self.Q
        return self.m[:2]  # Returns predicted position

    def update(self, y_obs):
        """Measurement Update Step: Corrects state estimate with new GPS observation."""
        # Innovation residual
        v = y_obs - (self.H @ self.m)
        
        # Innovation covariance
        S = (self.H @ self.P @ self.H.T) + self.R
        
        # Kalman Gain
        K = self.P @ self.H.T @ np.linalg.inv(S)
        
        # Posterior update
        self.m = self.m + (K @ v)
        self.P = (np.eye(4) - (K @ self.H)) @ self.P
        return self.m


# =====================================================================
# SIMULATION AND EVALUATION
# =====================================================================
if __name__ == "__main__":
    # Simulation Parameters
    np.random.seed(42)
    num_steps = 60
    delta_t = 1.0
    process_noise_std = 0.15   # Accel jitter variance
    gps_noise_std = 4.0        # High GPS variance (meters error)

    # 1. Generate Ground Truth Path (Constant velocity + small random accelerations)
    true_x, true_y = 0.0, 0.0
    true_vx, true_vy = 2.5, 4.0
    
    gt_positions = []
    gps_measurements = []
    
    for _ in range(num_steps):
        # Update ground truth positions
        true_x += true_vx * delta_t
        true_y += true_vy * delta_t
        gt_positions.append([true_x, true_y])
        
        # Add random acceleration perturbations to velocity
        true_vx += np.random.normal(0, process_noise_std) * delta_t
        true_vy += np.random.normal(0, process_noise_std) * delta_t
        
        # Simulate noisy sensor measurements
        meas_x = true_x + np.random.normal(0, gps_noise_std)
        meas_y = true_y + np.random.normal(0, gps_noise_std)
        gps_measurements.append([meas_x, meas_y])

    gt_positions = np.array(gt_positions)
    gps_measurements = np.array(gps_measurements)

    # 2. Run the Filter Over Generated Sequence
    kf = KalmanFilter2D(dt=delta_t, q_var=process_noise_std**2, r_var=gps_noise_std**2)
    kf.initialize_state(gps_measurements[0, 0], gps_measurements[0, 1])
    
    filtered_states = []
    for measurement in gps_measurements:
        kf.predict()
        updated_state = kf.update(measurement)
        filtered_states.append(updated_state.copy())
        
    filtered_states = np.array(filtered_states)

    # 3. Performance Metrics
    raw_gps_rmse = np.sqrt(np.mean(np.sum((gps_measurements - gt_positions)**2, axis=1)))
    kalman_rmse = np.sqrt(np.mean(np.sum((filtered_states[:, :2] - gt_positions)**2, axis=1)))
    print(f"Raw GPS Position RMSE: {raw_gps_rmse:.3f} meters")
    print(f"Kalman Filtered RMSE: {kalman_rmse:.3f} meters")

    # 4. Visualization Mapping
    plt.figure(figsize=(11, 7))
    plt.plot(gt_positions[:, 0], gt_positions[:, 1], 'g-', lw=2, label="Ground Truth (Actual Path)")
    plt.scatter(gps_measurements[:, 0], gps_measurements[:, 1], color='red', alpha=0.5, s=20, label="Noisy GPS Observations")
    plt.plot(filtered_states[:, 0], filtered_states[:, 1], 'b-o', markersize=4, lw=1.8, label="Kalman Filter Estimate")
    
    plt.title("2D Tracking Analysis: Noisy GPS Trajectory Smoothing", fontsize=12, fontweight='bold')
    plt.xlabel("X Coordinate (m)")
    plt.ylabel("Y Coordinate (m)")
    plt.grid(True, linestyle=':', alpha=0.6)
    plt.legend(loc="upper left")
    plt.axis("equal")
    plt.show()
